# Biohub L2 K-DS: DaughterNet scores for every exported candidate node

Scores all rows of `coords_exp39.parquet` (dataset `vitaliineibauer/biohub-l2-coords`: 1,012,255 candidate nodes of the
exp39 base over the 199 train videos; float voxel coordinates of the full-resolution frame) with the K-DN fold models
from the attached output of `vitaliineibauer/biohub-l2-kdn-train`.

* **OOF** (`KDS_ENSEMBLE=0`, default): each row is scored by the model of its video's fold (the fold model that never
  trained on that video; folds from the trainer's `splits_used.json`, else the embedded `splits.json`)
  -> `node_p_oof.parquet`. A routing audit (summary `oof_routing_audit`) leaves a video NaN if its fold model trained
  on it or on a video of its leak group (`splits.json` meta.leak_groups, 44b6 crops that share GT).
* **Ensemble** (`KDS_ENSEMBLE=1`): softmax averaged over all fold models, what the submission kernel does
  -> `node_p_ens.parquet`. `KDS_ENSEMBLE=both` produces both from ONE pass (every fold model runs once per sample;
  OOF = the video's fold model, ensemble = the mean), so the frames are read and prepared once.
* Output columns `stem, node_id, fold, p_cont, p_par, p_dau` (softmax in `tools.daughternet.CLASSES` order), one row per
  input row in input order, plus `node_p_<mode>_summary.json` (rows per fold, NaN count, timing, contract checks,
  `score_nodes` re-verification).
* Input = the DivNet contract of `tools/daughternet.prepare_sample` (lags -1..+2, XY pool 4, 16x32x32 crop, percentile
  50/99.5 clipped to [-0.5, 6], gaussian marker). Frames are read once per video (kdn_core.FrameReader) by a prefetch
  thread pool; samples are cropped with the contract's own functions and checked against `prepare_sample` in every
  video (1 sample from raw frames + 2 from the shared frames; any difference aborts).
* Two worker processes, one per T4 (`CUDA_VISIBLE_DEVICES` 0 / 1), videos split greedily by rows + frames; fp32 unless
  `KDS_AMP=1`. Per-video parts in `/kaggle/working/parts/<mode>/` make a rerun resume.
* Embedded sources (tools/divnet_infer.py, tools/daughternet.py, kdn_core.py, kds_core.py, splits.json) are sha256-guarded.

```python
%%writefile /kaggle/working/kds_src/tools/divnet_infer.py
"""Inference for the public ``giorgosi/biohub-divnet-v2`` checkpoint.

The dataset manifest is the contract: four image channels at lags -1, 0, +1,
+2 plus a Gaussian centre marker, XY mean pooling by four, a 16x32x32 crop,
percentile normalisation, and a small 3-D U-Net followed by global pooling.

``read_test_frame`` is supplied by the caller (the public pipeline notebook or
the local B20 scorer). This module raises on an incompatible checkpoint or
tensor shape; a silent geometric fallback would invalidate B20.
"""

from __future__ import annotations

import os
from pathlib import Path
from typing import Callable

import numpy as np
import torch


class _DivNetConvBlock(torch.nn.Module):
    def __init__(self, in_channels: int, out_channels: int) -> None:
        super().__init__()
        self.block = torch.nn.Sequential(
            torch.nn.Conv3d(in_channels, out_channels, 3, padding=1, bias=False),
            torch.nn.GroupNorm(min(8, out_channels), out_channels),
            torch.nn.SiLU(inplace=True),
            torch.nn.Conv3d(out_channels, out_channels, 3, padding=1, bias=False),
            torch.nn.GroupNorm(min(8, out_channels), out_channels),
            torch.nn.SiLU(inplace=True),
        )

    def forward(self, value: torch.Tensor) -> torch.Tensor:
        return self.block(value)


class DivNetMitosisClassifier(torch.nn.Module):
    """Architecture reconstructed exactly from the public checkpoint shapes."""

    def __init__(self, in_channels: int = 5, base_channels: int = 16) -> None:
        super().__init__()
        c = int(base_channels)
        self.enc1 = _DivNetConvBlock(in_channels, c)
        self.pool1 = torch.nn.MaxPool3d(2, 2)
        self.enc2 = _DivNetConvBlock(c, c * 2)
        self.pool2 = torch.nn.MaxPool3d(2, 2)
        self.enc3 = _DivNetConvBlock(c * 2, c * 4)
        self.pool3 = torch.nn.MaxPool3d(2, 2)
        self.bottleneck = _DivNetConvBlock(c * 4, c * 8)
        self.up3 = torch.nn.ConvTranspose3d(c * 8, c * 4, 2, 2)
        self.dec3 = _DivNetConvBlock(c * 8, c * 4)
        self.up2 = torch.nn.ConvTranspose3d(c * 4, c * 2, 2, 2)
        self.dec2 = _DivNetConvBlock(c * 4, c * 2)
        self.up1 = torch.nn.ConvTranspose3d(c * 2, c, 2, 2)
        self.dec1 = _DivNetConvBlock(c * 2, c)
        self.gap = torch.nn.AdaptiveAvgPool3d((1, 1, 1))
        self.head = torch.nn.Linear(c, 1)

    def forward(self, value: torch.Tensor) -> torch.Tensor:
        e1 = self.enc1(value)
        e2 = self.enc2(self.pool1(e1))
        e3 = self.enc3(self.pool2(e2))
        bottleneck = self.bottleneck(self.pool3(e3))
        d3 = self.dec3(torch.cat([self.up3(bottleneck), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.head(self.gap(d1).flatten(1))


def _checkpoint_candidates() -> list[Path]:
    root = Path(__file__).resolve().parent.parent
    explicit = os.environ.get("BIOHUB_DIVNET_CHECKPOINT", "").strip()
    candidates = [Path(explicit)] if explicit else []
    candidates.extend(
        [
            root / "external" / "giorgosi__biohub-divnet-v2" / "best_overall.pt",
            root / "external" / "giorgosi__biohub-divnet-v2" / "weights" / "divnet" / "best_overall.pt",
            Path("/kaggle/input/biohub-divnet-v2/best_overall.pt"),
            Path("/kaggle/input/biohub-divnet-v2/weights/divnet/best_overall.pt"),
            Path("/kaggle/input/datasets/giorgosi/biohub-divnet-v2/best_overall.pt"),
            Path("/kaggle/input/datasets/giorgosi/biohub-divnet-v2/weights/divnet/best_overall.pt"),
        ]
    )
    return candidates


def load_divnet_mitosis_model(checkpoint: Path | None = None) -> dict[str, object]:
    if not globals().get("DIVNET_VERIFY", True):
        raise RuntimeError("DivNet is disabled by configuration")
    path = Path(checkpoint) if checkpoint is not None else next(
        (item for item in _checkpoint_candidates() if item.is_file()), None
    )
    if path is None:
        checked = "\n".join(str(item) for item in _checkpoint_candidates())
        raise FileNotFoundError(f"DivNet checkpoint not found. Checked:\n{checked}")
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    payload = torch.load(path, map_location="cpu", weights_only=False)
    state = payload.get("model_state", payload) if isinstance(payload, dict) else payload
    config = payload.get("config", {}) if isinstance(payload, dict) else {}
    model_cfg = config.get("model", {}) if isinstance(config, dict) else {}
    model = DivNetMitosisClassifier(
        in_channels=int(model_cfg.get("in_channels", 5)),
        base_channels=int(model_cfg.get("base_channels", 16)),
    )
    model.load_state_dict(state, strict=True)
    model.to(device).eval()
    return {
        "model": model,
        "device": device,
        "torch": torch,
        "path": path.resolve(),
        "config": config,
        "prepared_frames": {},
    }


def _pool_xy(frame: np.ndarray, factor: int) -> np.ndarray:
    frame = np.asarray(frame)
    if factor <= 1:
        return frame.astype(np.float32, copy=False)
    z, y, x = frame.shape
    y2, x2 = (y // factor) * factor, (x // factor) * factor
    value = frame[:, :y2, :x2].astype(np.float32, copy=False)
    return value.reshape(z, y2 // factor, factor, x2 // factor, factor).mean(axis=(2, 4))


def _normalise(frame: np.ndarray, lo_pct: float, hi_pct: float, clip: tuple[float, float]) -> np.ndarray:
    value = np.asarray(frame, dtype=np.float32)
    lo, hi = np.percentile(value, [lo_pct, hi_pct])
    if not np.isfinite(lo) or not np.isfinite(hi) or hi <= lo:
        return np.zeros_like(value)
    return np.clip((value - float(lo)) / (float(hi) - float(lo)), clip[0], clip[1]).astype(np.float32)


def _crop_with_padding(frame: np.ndarray, centre: tuple[int, int, int], shape: tuple[int, int, int]) -> np.ndarray:
    result = np.zeros(shape, dtype=np.float32)
    source_slices: list[slice] = []
    target_slices: list[slice] = []
    for coordinate, size, limit in zip(centre, shape, frame.shape):
        start = coordinate - size // 2
        stop = start + size
        source_start, source_stop = max(0, start), min(limit, stop)
        target_start = source_start - start
        target_stop = target_start + max(0, source_stop - source_start)
        source_slices.append(slice(source_start, source_stop))
        target_slices.append(slice(target_start, target_stop))
    result[tuple(target_slices)] = frame[tuple(source_slices)]
    return result


def _marker(shape: tuple[int, int, int], sigma: tuple[float, float, float]) -> np.ndarray:
    axes = [np.arange(size, dtype=np.float32) - size // 2 for size in shape]
    zz, yy, xx = np.meshgrid(*axes, indexing="ij")
    distance = (zz / sigma[0]) ** 2 + (yy / sigma[1]) ** 2 + (xx / sigma[2]) ** 2
    return np.exp(-0.5 * distance).astype(np.float32)


def prepare_divnet_input(
    dataset: str,
    t: int,
    parent_node: dict[str, object],
    divnet_bundle: dict[str, object],
    frame_cache: dict[object, np.ndarray],
    read_frame: Callable[[str, int, dict], np.ndarray] | None = None,
) -> np.ndarray:
    """Build one ``(5, 16, 32, 32)`` sample from the manifest recipe."""
    cfg = divnet_bundle.get("config", {})
    input_cfg = cfg.get("input", {}) if isinstance(cfg, dict) else {}
    norm_cfg = cfg.get("normalization", {}) if isinstance(cfg, dict) else {}
    lags = tuple(int(value) for value in input_cfg.get("image_lags", (-1, 0, 1, 2)))
    pool_xy = int(input_cfg.get("pool_xy", 4))
    shape = (
        int(input_cfg.get("crop_z", 16)),
        int(input_cfg.get("crop_yx", 32)),
        int(input_cfg.get("crop_yx", 32)),
    )
    sigma = tuple(float(value) for value in input_cfg.get("marker_sigma", (1.5, 2.0, 2.0)))
    clip = tuple(float(value) for value in norm_cfg.get("clip", (-0.5, 6.0)))
    reader = read_frame or globals().get("read_test_frame")
    if reader is None:
        raise RuntimeError("DivNet needs a read_frame callback or a global read_test_frame")
    prepared = divnet_bundle.setdefault("prepared_frames", {})
    channels: list[np.ndarray] = []
    for lag in lags:
        frame_t = max(0, int(t) + lag)
        key = (dataset, frame_t)
        if key not in prepared:
            raw = reader(dataset, frame_t, frame_cache)
            pooled = _pool_xy(raw, pool_xy)
            prepared[key] = _normalise(
                pooled,
                float(norm_cfg.get("lo_pct", 50.0)),
                float(norm_cfg.get("hi_pct", 99.5)),
                clip,
            )
            frame_cache.pop(frame_t, None)
        centre = (
            int(round(float(parent_node["z"]))),
            int(round(float(parent_node["y"]) / pool_xy)),
            int(round(float(parent_node["x"]) / pool_xy)),
        )
        channels.append(_crop_with_padding(prepared[key], centre, shape))
    channels.append(_marker(shape, sigma))
    sample = np.stack(channels, axis=0)
    if sample.shape != (5, *shape) or not np.isfinite(sample).all():
        raise RuntimeError(f"Invalid DivNet input: shape={sample.shape}, finite={np.isfinite(sample).all()}")
    return sample


def divnet_score_division(
    dataset: str,
    t: int,
    parent_node: dict[str, object],
    daughter1: dict[str, object],
    daughter2: dict[str, object],
    divnet_bundle: dict[str, object],
    frame_cache: dict[object, np.ndarray],
    read_frame: Callable[[str, int, dict], np.ndarray] | None = None,
) -> float:
    """Return the parent division probability; daughter arguments preserve the public API."""
    del daughter1, daughter2
    sample = prepare_divnet_input(dataset, t, parent_node, divnet_bundle, frame_cache, read_frame)
    torch_mod = divnet_bundle["torch"]
    tensor = torch_mod.from_numpy(sample).unsqueeze(0).to(
        device=divnet_bundle["device"], dtype=torch_mod.float32
    )
    with torch_mod.inference_mode():
        logit = divnet_bundle["model"](tensor)
        probability = float(torch_mod.sigmoid(logit).item())
    if not np.isfinite(probability):
        raise RuntimeError("DivNet produced a non-finite probability")
    return probability


DIVNET_BUNDLE = None
```

```python
%%writefile /kaggle/working/kds_src/tools/daughternet.py
"""DaughterNet: 3-class node classifier (continuation / parent / daughter) for the L2 division stage.

Shared contract between training (Kaggle K-DN kernels), the submission kernel patch (--patch divinsert) and the laptop
replay. The input sample is EXACTLY the DivNet manifest contract of tools/divnet_infer.prepare_divnet_input: image
channels at lags -1, 0, +1, +2 around the node's frame t, XY mean-pool by 4, a 16x32x32 crop centred on the node,
percentile 50/99.5 normalisation clipped to [-0.5, 6], plus one gaussian centre marker -> (5, 16, 32, 32).
The network is the DivNet UNet-GAP body with a 3-logit head. Class order is fixed: CLASSES = (cont, parent, daughter).
  parent   = GT dividing node at the last frame before the split (label at its own t)
  daughter = GT daughter node at its first frame (label at its own t)
  cont     = annotated node with in-degree = out-degree = 1, >= 2 frames from any GT division
Scoring averages softmax over the loaded fold models (ensemble) or uses one fold model per video (OOF).
"""
from __future__ import annotations

from pathlib import Path
from typing import Callable, Sequence

import numpy as np
import torch

try:  # repo layout
    from tools.divnet_infer import _DivNetConvBlock, prepare_divnet_input
except Exception:  # embedded in a notebook cell after divnet_infer's code (names already global)
    _DivNetConvBlock = globals().get("_DivNetConvBlock")
    prepare_divnet_input = globals().get("prepare_divnet_input")

CLASSES = ("cont", "parent", "daughter")
DEFAULT_CFG = {
    "input": {"image_lags": [-1, 0, 1, 2], "pool_xy": 4, "crop_z": 16, "crop_yx": 32, "marker_sigma": [1.5, 2.0, 2.0]},
    "normalization": {"lo_pct": 50.0, "hi_pct": 99.5, "clip": [-0.5, 6.0]},
    "model": {"in_channels": 5, "base_channels": 16, "n_classes": 3},
}


class DaughterNet(torch.nn.Module):
    """DivNet UNet-GAP body (same layer names as DivNetMitosisClassifier) with an n_classes head."""

    def __init__(self, in_channels: int = 5, base_channels: int = 16, n_classes: int = 3) -> None:
        super().__init__()
        c = int(base_channels)
        self.enc1 = _DivNetConvBlock(in_channels, c)
        self.pool1 = torch.nn.MaxPool3d(2, 2)
        self.enc2 = _DivNetConvBlock(c, c * 2)
        self.pool2 = torch.nn.MaxPool3d(2, 2)
        self.enc3 = _DivNetConvBlock(c * 2, c * 4)
        self.pool3 = torch.nn.MaxPool3d(2, 2)
        self.bottleneck = _DivNetConvBlock(c * 4, c * 8)
        self.up3 = torch.nn.ConvTranspose3d(c * 8, c * 4, 2, 2)
        self.dec3 = _DivNetConvBlock(c * 8, c * 4)
        self.up2 = torch.nn.ConvTranspose3d(c * 4, c * 2, 2, 2)
        self.dec2 = _DivNetConvBlock(c * 4, c * 2)
        self.up1 = torch.nn.ConvTranspose3d(c * 2, c, 2, 2)
        self.dec1 = _DivNetConvBlock(c * 2, c)
        self.gap = torch.nn.AdaptiveAvgPool3d((1, 1, 1))
        self.head = torch.nn.Linear(c, int(n_classes))

    def forward(self, value: torch.Tensor) -> torch.Tensor:
        e1 = self.enc1(value)
        e2 = self.enc2(self.pool1(e1))
        e3 = self.enc3(self.pool2(e2))
        b = self.bottleneck(self.pool3(e3))
        d3 = self.dec3(torch.cat([self.up3(b), e3], dim=1))
        d2 = self.dec2(torch.cat([self.up2(d3), e2], dim=1))
        d1 = self.dec1(torch.cat([self.up1(d2), e1], dim=1))
        return self.head(self.gap(d1).flatten(1))


def build_model(cfg: dict | None = None) -> DaughterNet:
    m = (cfg or DEFAULT_CFG)["model"]
    return DaughterNet(int(m["in_channels"]), int(m["base_channels"]), int(m["n_classes"]))


def load_daughternet(paths: Sequence[Path | str], device: str | None = None) -> dict:
    """Load one or more fold checkpoints ({'model_state', 'config'}) -> bundle usable by prepare_sample/score_nodes."""
    dev = torch.device(device or ("cuda" if torch.cuda.is_available() else "cpu"))
    models, cfg = [], None
    for p in paths:
        payload = torch.load(Path(p), map_location="cpu", weights_only=False)
        cfg = payload.get("config", DEFAULT_CFG) if isinstance(payload, dict) else DEFAULT_CFG
        model = build_model(cfg)
        model.load_state_dict(payload["model_state"] if "model_state" in payload else payload, strict=True)
        models.append(model.to(dev).eval())
    if not models:
        raise FileNotFoundError("no DaughterNet checkpoints given")
    return {"models": models, "device": dev, "torch": torch, "config": cfg, "prepared_frames": {}}


def prepare_sample(dataset: str, t: int, node: dict, bundle: dict, frame_cache: dict,
                   read_frame: Callable[[str, int, dict], np.ndarray] | None = None) -> np.ndarray:
    """(5, 16, 32, 32) float32 sample, identical to the DivNet contract (bundle['config'] carries input/normalization)."""
    return prepare_divnet_input(dataset, int(t), node, bundle, frame_cache, read_frame)


@torch.no_grad()
def score_nodes(dataset: str, nodes: Sequence[dict], bundle: dict, frame_cache: dict,
                read_frame: Callable[[str, int, dict], np.ndarray] | None = None, batch: int = 64,
                model_index: int | None = None) -> np.ndarray:
    """Softmax probabilities (N, 3) in CLASSES order; mean over fold models unless model_index selects one (OOF)."""
    if not nodes:
        return np.zeros((0, 3), dtype=np.float32)
    order = sorted(range(len(nodes)), key=lambda i: int(nodes[i]["t"]))
    out = np.zeros((len(nodes), 3), dtype=np.float32)
    models = bundle["models"] if model_index is None else [bundle["models"][model_index]]
    dev = bundle["device"]
    for s in range(0, len(order), batch):
        idx = order[s:s + batch]
        x = np.stack([prepare_sample(dataset, int(nodes[i]["t"]), nodes[i], bundle, frame_cache, read_frame) for i in idx])
        xt = torch.from_numpy(x).to(dev, dtype=torch.float32)
        p = torch.stack([torch.softmax(m(xt), dim=1) for m in models]).mean(0)
        out[idx] = p.float().cpu().numpy()
    return out
```

```python
%%writefile /kaggle/working/kds_src/kdn_core.py
"""K-DN: train DaughterNet (tools/daughternet.py) from scratch with honest video folds and write out-of-fold predictions.

One module shared by the Kaggle kernel (kdn-train.ipynb materialises it with %%writefile next to tools/divnet_infer.py
and tools/daughternet.py) and by the laptop runner run_local.py. Steps (run_all runs them in order):

  step_prepare      locate the data, folds (embedded splits.json, else the inline seed-0 rule), GT labels per video
  step_build_cache  DivNet-contract crops for every labelled node -> one memmap (N, 4, 18, 34, 34) + meta.csv
  step_train        one subprocess per fold (two at once on 2 GPUs): balanced sampler, flips/rot90/jitter/gain,
                    AdamW + cosine, early stopping on the fold's val loss, best.pt/last.pt every epoch, fold OOF
  step_oof          oof.csv + oof_readout.json (+ in-sample DivNet comparison)

Labels (annotated GT nodes only, never unannotated space):
  parent   node with >= 2 GT children (at its own t)
  daughter each child of a parent (at its own, first, frame)
  cont     in-degree 1, out-degree 1, >= cont_min_dist frames (lineage) from every parent AND daughter node,
           random subsample of cont_per_video per video
  hardneg  cont nodes that sit in the L2 insertion gate of an annotated one-child GT node, as u or as v
           (u: exactly one child c1 at t+1 within 10 um; v at t+1, v != c1, within 15 um of u; 3 <= |c1-v| <= 20 um;
           |midpoint(c1, v) - u| <= 6.5 um; cos(c1-u, v-u) <= 0.25), up to hardneg_per_video, label cont
Crops: prepare_sample (DivNet contract) called with an enlarged 18x34x34 crop; the inner 16x32x32 of it equals the
contract sample exactly (asserted on the first nodes of every video), the 1-voxel margin feeds the +-1 voxel jitter.
Node coordinates are voxel indices of the full-resolution (64, 256, 256) frame, exactly as in the GT .geff.
Frame index t+lag is clamped to [0, T-1] (the contract clamps the low end with max(0, .); the high end is clamped here,
as in friend A's FrameReader, so nodes at t >= T-2 get a repeated last frame).
"""
from __future__ import annotations

import copy
import json
import math
import os
import struct
import subprocess
import sys
import threading
import time
import zlib
from pathlib import Path

import numpy as np

try:  # repo layout (laptop) and the Kaggle kernel (tools/ materialised under kdn_src)
    from tools.daughternet import CLASSES, DEFAULT_CFG, build_model, load_daughternet, prepare_sample, score_nodes
    from tools.divnet_infer import _marker, load_divnet_mitosis_model, _checkpoint_candidates
except Exception:  # pragma: no cover - notebook fallback when the cells were executed into globals
    CLASSES = globals()["CLASSES"]
    DEFAULT_CFG = globals()["DEFAULT_CFG"]
    build_model = globals()["build_model"]
    load_daughternet = globals()["load_daughternet"]
    prepare_sample = globals()["prepare_sample"]
    score_nodes = globals()["score_nodes"]
    _marker = globals()["_marker"]
    load_divnet_mitosis_model = globals()["load_divnet_mitosis_model"]
    _checkpoint_candidates = globals().get("_checkpoint_candidates", lambda: [])

COMPETITION = "biohub-cell-tracking-during-development"
VOXEL_SCALE_UM = np.array([1.625, 0.40625, 0.40625], dtype=np.float64)
CLASS_INDEX = {name: i for i, name in enumerate(CLASSES)}
CROP = (16, 32, 32)          # DivNet contract crop (z, y, x) in pooled voxels
MARGIN = 1                   # +-1 voxel jitter margin stored around it
BIG = (CROP[0] + 2 * MARGIN, CROP[1] + 2 * MARGIN, CROP[2] + 2 * MARGIN)
OOF_COLUMNS = ["stem", "t", "z", "y", "x", "label", "fold", "p_cont", "p_par", "p_dau"]


def log(*args) -> None:
    print(time.strftime("%H:%M:%S"), *args, flush=True)


# ----------------------------------------------------------------------------------------------------------------------
# configuration
# ----------------------------------------------------------------------------------------------------------------------
def _env_flag(name: str, default: str = "0") -> bool:
    return os.environ.get(name, default).strip() not in ("", "0", "false", "False", "no")


def config_from_env(overrides: dict | None = None) -> dict:
    env = os.environ.get
    smoke = _env_flag("KDN_SMOKE")
    on_kaggle = Path("/kaggle/working").exists()
    cfg = {
        "smoke": smoke,
        "data_dir": env("KDN_DATA_DIR", ""),
        "work_dir": env("KDN_WORK_DIR", "/kaggle/working" if on_kaggle else str(Path.cwd() / "kdn_out")),
        "cache_dir": env("KDN_CACHE_DIR", ""),
        "splits_path": env("KDN_SPLITS", ""),
        "stems": [s.strip() for s in env("KDN_STEMS", "").split(",") if s.strip()],
        "max_videos": int(env("KDN_MAX_VIDEOS", "0") or 0),
        "smoke_videos": int(env("KDN_SMOKE_VIDEOS", "5")),
        "n_folds": int(env("KDN_FOLDS", "2" if smoke else "5")),
        "epochs": int(env("KDN_EPOCHS", "1" if smoke else "30")),
        "patience": int(env("KDN_PATIENCE", "6")),
        "batch": int(env("KDN_BATCH", "64")),
        "lr": float(env("KDN_LR", "1e-3")),
        "weight_decay": float(env("KDN_WD", "0.01")),
        "warmup_frac": float(env("KDN_WARMUP_FRAC", "0.03")),
        "epoch_samples": int(env("KDN_EPOCH_SAMPLES", "0")),   # 0 = number of training samples
        "eval_batch": int(env("KDN_EVAL_BATCH", "256")),
        "cont_per_video": int(env("KDN_CONT_PER_VIDEO", "60")),
        "hardneg_per_video": int(env("KDN_HARDNEG_PER_VIDEO", "40")),
        "cont_min_dist": int(env("KDN_CONT_MIN_DIST", "2")),
        "seed": int(env("KDN_SEED", "0")),
        "build_workers": int(env("KDN_BUILD_WORKERS", "0")),  # 0 = min(4, cpu count)
        "contract_checks": int(env("KDN_CONTRACT_CHECKS", "2")),  # exact contract checks per video
        "cache_dtype": env("KDN_CACHE_DTYPE", "float32"),
        "jitter": int(env("KDN_JITTER", "1")),
        "gain": [float(env("KDN_GAIN_LO", "0.8")), float(env("KDN_GAIN_HI", "1.2"))],
        "amp": _env_flag("KDN_AMP", "1"),
        "gpus": int(env("KDN_GPUS", "-1")),                   # -1 = all visible, 0 = CPU
        "inner_val": _env_flag("KDN_INNER_VAL"),              # 1 = early-stop on fold k+1, train on the other K-2
        "divnet_compare": _env_flag("KDN_DIVNET", "1"),
        "divnet_checkpoint": env("KDN_DIVNET_CKPT", ""),
        "keep_cache": _env_flag("KDN_KEEP_CACHE", "1"),
        "gate": {"child_um": 10.0, "v_um": 15.0, "sister_min_um": 3.0, "sister_max_um": 20.0,
                 "mid_um": 6.5, "cos_max": 0.25},
    }
    if overrides:
        cfg.update(overrides)
    return cfg


def resolve_data_dir(cfg: dict) -> Path:
    if cfg.get("data_dir"):
        return Path(cfg["data_dir"])
    candidates = [Path(f"/kaggle/input/competitions/{COMPETITION}"), Path(f"/kaggle/input/{COMPETITION}")]
    comp_dir = next((p for p in candidates if p.exists()), candidates[0])
    return comp_dir / "train"


def resolve_cache_dir(cfg: dict) -> Path:
    if cfg.get("cache_dir"):
        return Path(cfg["cache_dir"])
    for base in (Path("/kaggle/temp"), Path("/tmp")):
        if base.exists() and os.access(base, os.W_OK):
            return base / "kdn_cache"
    return Path(cfg["work_dir"]) / "_kdn_cache"


# ----------------------------------------------------------------------------------------------------------------------
# codecs: zstd (GT .geff arrays) and blosc (image chunks); several decoders so the kernel needs no extra wheels
# ----------------------------------------------------------------------------------------------------------------------
_ZSTD_FN = None


def _zstd_candidates():
    def via_zstandard(buf, size):
        import zstandard
        return zstandard.ZstdDecompressor().decompress(buf, max_output_size=max(size, 1))

    def via_numcodecs(buf, size):
        from numcodecs import Zstd
        return bytes(Zstd().decode(buf))

    def via_pyarrow(buf, size):
        import pyarrow as pa
        return pa.Codec("zstd").decompress(buf, decompressed_size=size, asbytes=True)

    def via_imagecodecs(buf, size):
        import imagecodecs
        return bytes(imagecodecs.zstd_decode(buf))

    return [via_zstandard, via_numcodecs, via_pyarrow, via_imagecodecs]


def zstd_decompress(buf: bytes, size: int) -> bytes:
    global _ZSTD_FN
    if _ZSTD_FN is not None:
        return _ZSTD_FN(buf, size)
    errors = []
    for fn in _zstd_candidates():
        try:
            out = fn(buf, size)
            if len(out) != size:
                raise ValueError(f"size {len(out)} != {size}")
            _ZSTD_FN = fn
            return out
        except Exception as exc:  # try the next decoder
            errors.append(f"{fn.__name__}: {type(exc).__name__}: {exc}")
    raise RuntimeError("no zstd decoder works: " + " | ".join(errors))


def _bitunshuffle(block: bytes, typesize: int) -> bytes:
    n = len(block) // typesize
    n8 = n - (n % 8)
    if n8 == 0:
        return block
    head = np.frombuffer(block, dtype=np.uint8, count=n8 * typesize).reshape(typesize * 8, n8 // 8)
    bits = np.unpackbits(head, axis=1, bitorder="little").reshape(typesize, 8, n8)
    elems = np.packbits(bits, axis=1, bitorder="little").reshape(typesize, n8)
    return elems.T.tobytes() + block[n8 * typesize:]


def blosc1_decompress_pure(raw: bytes) -> bytes:
    """Blosc-1 frame decoder (numpy + a zstd decoder); last-resort fallback, tested against blosc2 locally."""
    flags, typesize = raw[2], raw[3]
    nbytes, blocksize, _cbytes = struct.unpack("<iii", raw[4:16])
    if flags & 0x2:  # memcpyed
        return bytes(raw[16:16 + nbytes])
    compcode = flags >> 5
    if compcode != 4:
        raise RuntimeError(f"pure blosc decoder supports zstd only (compcode {compcode})")
    nblocks = (nbytes + blocksize - 1) // blocksize
    bstarts = struct.unpack(f"<{nblocks}i", raw[16:16 + 4 * nblocks])
    out = bytearray(nbytes)
    dont_split = bool(flags & 0x10)
    for i, start in enumerate(bstarts):
        bsize = min(blocksize, nbytes - i * blocksize)
        nstreams = 1 if (dont_split or typesize <= 1 or bsize < typesize) else typesize
        neb = bsize // nstreams
        pos, parts = start, []
        for _ in range(nstreams):
            csize = struct.unpack("<i", raw[pos:pos + 4])[0]
            pos += 4
            chunk = raw[pos:pos + csize]
            pos += csize
            parts.append(bytes(chunk) if csize == neb else zstd_decompress(bytes(chunk), neb))
        block = b"".join(parts)
        if flags & 0x4:
            block = _bitunshuffle(block, typesize)
        elif (flags & 0x1) and typesize > 1:
            n = bsize // typesize
            block = np.frombuffer(block, np.uint8, count=n * typesize).reshape(typesize, n).T.tobytes() + block[n * typesize:]
        out[i * blocksize:i * blocksize + bsize] = block
    return bytes(out)


_BLOSC_FN = None


def _blosc_candidates():
    def via_blosc2(raw):
        import blosc2
        return blosc2.decompress(raw)

    def via_numcodecs(raw):
        from numcodecs import Blosc
        return bytes(Blosc().decode(raw))

    def via_blosc(raw):
        import blosc
        return blosc.decompress(raw)

    return [via_blosc2, via_numcodecs, via_blosc, blosc1_decompress_pure]


def blosc_decompress(raw: bytes) -> bytes:
    global _BLOSC_FN
    nbytes = struct.unpack("<i", raw[4:8])[0]
    if _BLOSC_FN is not None:
        return _BLOSC_FN(raw)
    errors = []
    for fn in _blosc_candidates():
        try:
            out = fn(raw)
            if len(out) != nbytes:
                raise ValueError(f"size {len(out)} != {nbytes}")
            _BLOSC_FN = fn
            return out
        except Exception as exc:
            errors.append(f"{fn.__name__}: {type(exc).__name__}: {exc}")
    raise RuntimeError("no blosc decoder works: " + " | ".join(errors))


def decoder_names() -> dict:
    return {"zstd": getattr(_ZSTD_FN, "__name__", None), "blosc": getattr(_BLOSC_FN, "__name__", None)}


def _zarr_dtype(meta: dict) -> np.dtype:
    dtype = np.dtype(meta["data_type"])
    for codec in meta.get("codecs", []):
        if codec.get("name") == "bytes":
            endian = (codec.get("configuration") or {}).get("endian", "little")
            dtype = dtype.newbyteorder("<" if endian == "little" else ">")
    return dtype


def _decode_chunk(raw: bytes, meta: dict, chunk_shape: tuple) -> np.ndarray:
    dtype = _zarr_dtype(meta)
    size = int(np.prod(chunk_shape)) * dtype.itemsize
    data = raw
    codecs = meta.get("codecs", [])
    for codec in reversed(codecs):
        name = codec.get("name")
        if name == "bytes":
            continue
        if name == "zstd":
            data = zstd_decompress(data, size)
        elif name == "blosc":
            data = blosc_decompress(data)
        elif name == "gzip":
            data = zlib.decompress(data, 16 + zlib.MAX_WBITS)
        else:
            raise RuntimeError(f"unsupported zarr codec {name}")
    return np.frombuffer(data, dtype=dtype, count=int(np.prod(chunk_shape))).reshape(chunk_shape)


def read_zarr_array(path: Path) -> np.ndarray:
    """Whole zarr v3 array with a regular chunk grid, decoded with numpy (+ zstd/blosc); zarr is the fallback."""
    path = Path(path)
    try:
        meta = json.loads((path / "zarr.json").read_text())
        shape = tuple(int(v) for v in meta["shape"])
        chunks = tuple(int(v) for v in meta["chunk_grid"]["configuration"]["chunk_shape"])
        sep = ((meta.get("chunk_key_encoding") or {}).get("configuration") or {}).get("separator", "/")
        dtype = _zarr_dtype(meta)
        out = np.full(shape, meta.get("fill_value", 0) or 0, dtype=dtype.newbyteorder("="))
        grid = [range(math.ceil(s / c)) if s else range(0) for s, c in zip(shape, chunks)]
        for idx in np.ndindex(*[len(g) for g in grid]):
            key = sep.join(str(i) for i in idx)
            chunk_path = path / "c" / key if sep == "/" else path / f"c{sep}{key}"
            if not chunk_path.exists():
                continue
            chunk = _decode_chunk(chunk_path.read_bytes(), meta, chunks)
            sl = tuple(slice(i * c, min((i + 1) * c, s)) for i, c, s in zip(idx, chunks, shape))
            out[sl] = chunk[tuple(slice(0, x.stop - x.start) for x in sl)]
        return out
    except Exception as first:
        try:
            import zarr
            return np.asarray(zarr.open(str(path), mode="r")[...])
        except Exception:
            raise first


def read_gt_geff(path: Path) -> dict:
    """GT graph arrays: ids, t, z, y, x (voxel units of the full-resolution frame) and edges (E, 2) source->target."""
    path = Path(path)
    ids = read_zarr_array(path / "nodes" / "ids").astype(np.int64)
    props = path / "nodes" / "props"
    cols = {k: read_zarr_array(props / k / "values") for k in ("t", "z", "y", "x")}
    edges_path = path / "edges" / "ids"
    edges = read_zarr_array(edges_path).astype(np.int64).reshape(-1, 2) if edges_path.exists() else np.zeros((0, 2), np.int64)
    return {"ids": ids, "t": cols["t"].astype(np.int64), "z": cols["z"].astype(np.float64),
            "y": cols["y"].astype(np.float64), "x": cols["x"].astype(np.float64), "edges": edges}


class FrameReader:
    """read_test_frame of the exp39 notebook, parameterised by directory: one blosc chunk per frame, zarr fallback.

    Callable as read_frame(dataset, t, frame_cache); t is clamped to [0, T-1]; the frame is cached under the requested
    t so prepare_divnet_input's frame_cache.pop(frame_t) releases it."""

    def __init__(self, data_dir: Path) -> None:
        self.data_dir = Path(data_dir)
        self.meta: dict[str, tuple] = {}
        self.frames_read = 0

    def _meta(self, dataset: str) -> tuple:
        if dataset not in self.meta:
            meta = json.loads((self.data_dir / f"{dataset}.zarr" / "0" / "zarr.json").read_text())
            shape = tuple(int(v) for v in meta["shape"])
            chunks = tuple(int(v) for v in meta["chunk_grid"]["configuration"]["chunk_shape"])
            self.meta[dataset] = (meta, shape, chunks)
        return self.meta[dataset]

    def __call__(self, dataset: str, t: int, frame_cache: dict) -> np.ndarray:
        if t in frame_cache:
            return frame_cache[t]
        meta, shape, chunks = self._meta(dataset)
        index = max(0, min(int(t), shape[0] - 1))
        zarr_path = self.data_dir / f"{dataset}.zarr"
        frame = None
        if chunks[0] == 1 and tuple(chunks[1:]) == tuple(shape[1:]):
            try:
                raw = (zarr_path / "0" / "c" / str(index) / "0" / "0" / "0").read_bytes()
                frame = _decode_chunk(raw, meta, (1, *shape[1:]))[0].copy()
            except Exception:
                frame = None
        if frame is None:
            import zarr
            frame = np.asarray(zarr.open(str(zarr_path / "0"), mode="r")[index])
        self.frames_read += 1
        frame_cache[t] = frame
        return frame


# ----------------------------------------------------------------------------------------------------------------------
# labels
# ----------------------------------------------------------------------------------------------------------------------
def _stem_seed(stem: str, seed: int) -> list[int]:
    return [int(seed), zlib.crc32(stem.encode("utf-8"))]


def extract_labels(stem: str, geff_path: Path, cfg: dict) -> tuple[list[dict], dict]:
    g = read_gt_geff(geff_path)
    ids = g["ids"]
    n = len(ids)
    pos_of = {int(v): i for i, v in enumerate(ids)}
    t = g["t"]
    pts_um = np.stack([g["z"], g["y"], g["x"]], axis=1) * VOXEL_SCALE_UM
    children: list[list[int]] = [[] for _ in range(n)]
    parents: list[list[int]] = [[] for _ in range(n)]
    dropped_edges = 0
    for s, d in g["edges"]:
        a, b = pos_of.get(int(s)), pos_of.get(int(d))
        if a is None or b is None:
            dropped_edges += 1
            continue
        children[a].append(b)
        parents[b].append(a)
    outdeg = np.array([len(c) for c in children])
    indeg = np.array([len(p) for p in parents])
    is_parent = outdeg >= 2
    is_daughter = np.zeros(n, bool)
    for i in np.flatnonzero(is_parent):
        is_daughter[children[i]] = True
    ambiguous = is_parent & is_daughter
    division_node = is_parent | is_daughter

    # lineage distance (frames) to the nearest parent or daughter node among ancestors and descendants
    order = np.argsort(t, kind="stable")
    up = np.full(n, -np.inf)
    for i in order:
        if division_node[i]:
            up[i] = t[i]
        elif parents[i]:
            up[i] = max(up[p] for p in parents[i])
    down = np.full(n, np.inf)
    for i in order[::-1]:
        if division_node[i]:
            down[i] = t[i]
        elif children[i]:
            down[i] = min(down[c] for c in children[i])
    dist = np.minimum(t - up, down - t)
    cont_ok = (indeg == 1) & (outdeg == 1) & ~division_node & (dist >= int(cfg["cont_min_dist"]))

    # hard negatives: cont nodes in the insertion gate of an annotated one-child node (as u or as v)
    gate = cfg["gate"]
    by_t: dict[int, np.ndarray] = {}
    for tt in np.unique(t):
        by_t[int(tt)] = np.flatnonzero(t == tt)
    hard_role: dict[int, set] = {}
    gate_triples = 0
    for u in np.flatnonzero(outdeg == 1):
        c1 = children[u][0]
        if t[c1] != t[u] + 1:
            continue
        vec_c = pts_um[c1] - pts_um[u]
        n_c = float(np.linalg.norm(vec_c))
        if n_c > gate["child_um"]:
            continue
        cand = by_t.get(int(t[u]) + 1)
        if cand is None:
            continue
        cand = cand[cand != c1]
        vec_v = pts_um[cand] - pts_um[u]
        n_v = np.linalg.norm(vec_v, axis=1)
        sister = np.linalg.norm(pts_um[cand] - pts_um[c1], axis=1)
        mid = np.linalg.norm(0.5 * (pts_um[cand] + pts_um[c1]) - pts_um[u], axis=1)
        cos = (vec_v @ vec_c) / np.maximum(n_c * n_v, 1e-9)  # zero-length vector -> cos 0 (passes)
        ok = ((n_v <= gate["v_um"]) & (sister >= gate["sister_min_um"]) & (sister <= gate["sister_max_um"])
              & (mid <= gate["mid_um"]) & (cos <= gate["cos_max"]))
        for v in cand[ok]:
            gate_triples += 1
            if cont_ok[u]:
                hard_role.setdefault(int(u), set()).add("u")
            if cont_ok[v]:
                hard_role.setdefault(int(v), set()).add("v")

    rng = np.random.default_rng(_stem_seed(stem, cfg["seed"]))
    hard_nodes = sorted(hard_role)
    if len(hard_nodes) > cfg["hardneg_per_video"]:
        hard_nodes = sorted(rng.choice(hard_nodes, size=cfg["hardneg_per_video"], replace=False).tolist())
    hard_set = set(hard_nodes)
    cont_pool = [i for i in np.flatnonzero(cont_ok) if int(i) not in hard_set]
    if len(cont_pool) > cfg["cont_per_video"]:
        cont_pool = sorted(rng.choice(cont_pool, size=cfg["cont_per_video"], replace=False).tolist())

    rows: list[dict] = []

    def add(i: int, label: str, kind: str) -> None:
        rows.append({"stem": stem, "node_id": int(ids[i]), "t": int(t[i]), "z": float(g["z"][i]),
                     "y": float(g["y"][i]), "x": float(g["x"][i]), "label": label, "kind": kind})

    for i in np.flatnonzero(is_parent & ~ambiguous):
        add(int(i), "parent", "parent")
    for i in np.flatnonzero(is_daughter & ~ambiguous):
        add(int(i), "daughter", "daughter")
    for i in cont_pool:
        add(int(i), "cont", "cont")
    for i in hard_nodes:
        add(int(i), "cont", "hardneg_" + "".join(sorted(hard_role[i])))
    stats = {
        "stem": stem, "embryo": stem[:4], "gt_nodes": int(n), "gt_edges": int(len(g["edges"])),
        "dropped_edges": int(dropped_edges), "divisions": int(is_parent.sum()),
        "parent": int((is_parent & ~ambiguous).sum()), "daughter": int((is_daughter & ~ambiguous).sum()),
        "ambiguous_dropped": int(ambiguous.sum()), "cont_eligible": int(cont_ok.sum()),
        "cont": len(cont_pool), "gate_triples": int(gate_triples), "hardneg_candidates": len(hard_role),
        "hardneg": len(hard_nodes),
        "hardneg_u": sum(1 for i in hard_nodes if "u" in hard_role[i]),
        "hardneg_v": sum(1 for i in hard_nodes if "v" in hard_role[i]),
        "edge_dt_not_1": int(sum(1 for a in range(n) for b in children[a] if t[b] - t[a] != 1)),
    }
    return rows, stats


# ----------------------------------------------------------------------------------------------------------------------
# folds
# ----------------------------------------------------------------------------------------------------------------------
def make_splits_inline(n_div: dict[str, int], n_folds: int = 5, seed: int = 0) -> dict[str, int]:
    """Fallback only (used when no splits.json is available): 5 folds grouped by video, stratified by embryo prefix
    and division count, seed 0. Per embryo: videos in seeded random order, stable-sorted by division count (desc);
    division videos go greedily to the fold with the fewest divisions (then fewest videos), zero-division videos to
    the fold with the fewest videos of that embryo."""
    rng = np.random.default_rng(seed)
    folds: dict[str, int] = {}
    for emb in sorted({s[:4] for s in n_div}):
        vids = sorted(s for s in n_div if s[:4] == emb)
        vids = [vids[i] for i in rng.permutation(len(vids))]
        vids.sort(key=lambda s: -n_div[s])
        div_sum, cnt = [0] * n_folds, [0] * n_folds
        for s in vids:
            if n_div[s] > 0:
                f = min(range(n_folds), key=lambda k: (div_sum[k], cnt[k], k))
            else:
                f = min(range(n_folds), key=lambda k: (cnt[k], k))
            folds[s] = f
            div_sum[f] += n_div[s]
            cnt[f] += 1
    return folds


def load_splits(path: Path) -> dict[str, int] | None:
    if not path or not Path(path).is_file():
        return None
    d = json.loads(Path(path).read_text())
    if isinstance(d, dict) and isinstance(d.get("folds"), dict):
        return {str(k): int(v) for k, v in d["folds"].items()}
    if isinstance(d, dict) and isinstance(d.get("folds"), list):
        return {str(s): i for i, group in enumerate(d["folds"]) for s in group}
    if isinstance(d, dict) and all(isinstance(v, int) for v in d.values()):
        return {str(k): int(v) for k, v in d.items()}
    if isinstance(d, list):
        return {str(r["stem"]): int(r["fold"]) for r in d}
    raise ValueError(f"unrecognised splits format in {path}")


def _default_splits_path() -> Path | None:
    here = Path(__file__).resolve().parent
    for p in (here / "splits.json", here.parent / "splits.json"):
        if p.is_file():
            return p
    return None


# ----------------------------------------------------------------------------------------------------------------------
# step 1: prepare
# ----------------------------------------------------------------------------------------------------------------------
def step_prepare(cfg: dict) -> dict:
    import pandas as pd

    t0 = time.time()
    data_dir = resolve_data_dir(cfg)
    work = Path(cfg["work_dir"])
    work.mkdir(parents=True, exist_ok=True)
    cache = resolve_cache_dir(cfg)
    cache.mkdir(parents=True, exist_ok=True)
    all_stems = sorted(p.name[:-5] for p in data_dir.iterdir() if p.name.endswith(".geff")
                       and (data_dir / f"{p.name[:-5]}.zarr").exists())
    log(f"[prepare] data_dir={data_dir} videos with .zarr+.geff: {len(all_stems)}; cache={cache}; work={work}")
    if not all_stems:
        raise FileNotFoundError(f"no <stem>.zarr + <stem>.geff pairs under {data_dir}")

    # GT labels for every video (cheap: the .geff files are tiny); per-video try/except
    label_rows, stats_rows, n_div = [], [], {}
    for i, stem in enumerate(all_stems):
        ts = time.time()
        try:
            rows, st = extract_labels(stem, data_dir / f"{stem}.geff", cfg)
            st["label_secs"] = round(time.time() - ts, 3)
            label_rows.append(rows)
            stats_rows.append(st)
            n_div[stem] = st["divisions"]
        except Exception as exc:
            log(f"[prepare] {stem}: label extraction FAILED {type(exc).__name__}: {exc}")
            stats_rows.append({"stem": stem, "embryo": stem[:4], "error": f"{type(exc).__name__}: {exc}"})
            label_rows.append([])
            n_div[stem] = 0
    log(f"[prepare] GT parsed for {len(all_stems)} videos in {time.time() - t0:.1f} s; "
        f"divisions {sum(n_div.values())} in {sum(1 for v in n_div.values() if v)} videos; zstd decoder {decoder_names()['zstd']}")

    # folds
    splits_path = Path(cfg["splits_path"]) if cfg.get("splits_path") else _default_splits_path()
    splits = load_splits(splits_path) if splits_path else None
    splits_source = str(splits_path) if splits else "inline make_splits_inline(seed=0)"
    if splits is None:
        splits = make_splits_inline(n_div, 5, 0)
    missing = [s for s in all_stems if s not in splits]
    if missing:
        log(f"[prepare] WARNING {len(missing)} stems missing from splits -> inline fold rule for them")
        extra = make_splits_inline({s: n_div[s] for s in missing}, 5, 0)
        splits.update(extra)

    # video selection
    if cfg["stems"]:
        stems = [s for s in cfg["stems"] if s in set(all_stems)]
    else:
        limit = cfg["max_videos"] or 0
        if cfg["smoke"]:
            limit = min(limit, cfg["smoke_videos"]) if limit else cfg["smoke_videos"]
        if limit and limit < len(all_stems):
            by_emb = {e: sorted((s for s in all_stems if s[:4] == e), key=lambda s: (-n_div[s], s))
                      for e in sorted({s[:4] for s in all_stems})}
            stems, k = [], 0
            while len(stems) < limit:  # alternate embryos, division-rich videos first
                emb = sorted(by_emb, reverse=True)[k % len(by_emb)]
                if by_emb[emb]:
                    stems.append(by_emb[emb].pop(0))
                k += 1
                if not any(by_emb.values()):
                    break
            stems = sorted(stems)
        else:
            stems = list(all_stems)
    n_folds = int(cfg["n_folds"])
    orig_folds = sorted({splits[s] for s in splits})
    if n_folds == len(orig_folds) and len(stems) == len(all_stems):
        folds = {s: int(splits[s]) for s in stems}
    else:
        folds = {s: int(splits[s]) % n_folds for s in stems}
        if len({folds[s] for s in stems}) < min(n_folds, len(stems)):  # small subsets: deal round-robin
            ordered = sorted(stems, key=lambda s: (splits[s], s))
            folds = {s: i % n_folds for i, s in enumerate(ordered)}

    chosen = set(stems)
    rows = [r for stem, rs in zip(all_stems, label_rows) if stem in chosen for r in rs]
    meta = pd.DataFrame(rows, columns=["stem", "node_id", "t", "z", "y", "x", "label", "kind"])
    meta["label_idx"] = meta["label"].map(CLASS_INDEX).astype(int)
    meta["fold"] = meta["stem"].map(folds).astype(int)
    meta["embryo"] = meta["stem"].str[:4]
    meta = meta.sort_values(["stem", "t", "node_id", "kind"]).reset_index(drop=True)
    stats = pd.DataFrame(stats_rows)
    stats["selected"] = stats["stem"].isin(stems)
    stats["fold"] = stats["stem"].map(folds)
    (work / "splits_used.json").write_text(json.dumps(
        {"source": splits_source, "n_folds": n_folds, "folds": {s: folds[s] for s in stems}}, indent=1))
    log(f"[prepare] folds from {splits_source}; selected {len(stems)} videos, {len(meta)} samples; "
        f"per fold videos {dict(sorted(pd.Series(folds).value_counts().items()))}")
    _print_label_counts(meta)
    return {"data_dir": str(data_dir), "cache_dir": str(cache), "work_dir": str(work), "stems": stems,
            "folds": folds, "meta": meta, "stats": stats, "splits_source": splits_source,
            "timing": {"prepare_s": round(time.time() - t0, 1)}}


def _print_label_counts(meta) -> None:
    if meta.empty:
        log("[prepare] no labelled samples")
        return
    tab = meta.groupby(["embryo", "kind"]).size().unstack(fill_value=0)
    log("[prepare] samples by embryo x kind:\n" + tab.to_string())
    tab = meta.groupby(["fold", "label"]).size().unstack(fill_value=0)
    log("[prepare] samples by fold x label:\n" + tab.to_string())


# ----------------------------------------------------------------------------------------------------------------------
# step 2: crops
# ----------------------------------------------------------------------------------------------------------------------
def _big_cfg() -> dict:
    cfg = copy.deepcopy(DEFAULT_CFG)
    cfg["input"]["crop_z"] = BIG[0]
    cfg["input"]["crop_yx"] = BIG[1]
    return cfg


def _build_worker_init(n_threads: int) -> None:
    try:
        import blosc2
        blosc2.set_nthreads(max(1, n_threads))
    except Exception:
        pass
    try:
        import torch
        torch.set_num_threads(1)
    except Exception:
        pass


def crop_video(job: dict) -> dict:
    """All crops of one video: frames are read once (sorted by t), prepared frames kept for lags -1..+2 only."""
    t0 = time.time()
    stem = job["stem"]
    try:
        reader = FrameReader(Path(job["data_dir"]))
        prepared: dict = {}
        bundle = {"config": _big_cfg(), "prepared_frames": prepared}
        ref_bundle = {"config": copy.deepcopy(DEFAULT_CFG), "prepared_frames": prepared}
        frame_cache: dict = {}
        nodes = job["nodes"]
        out = np.empty((len(nodes), 4, *BIG), dtype=np.dtype(job["dtype"]))
        checked = 0
        inner = (slice(None), slice(MARGIN, MARGIN + CROP[0]), slice(MARGIN, MARGIN + CROP[1]),
                 slice(MARGIN, MARGIN + CROP[2]))
        for i in sorted(range(len(nodes)), key=lambda k: nodes[k][0]):
            tt, z, y, x = nodes[i]
            node = {"t": int(tt), "z": float(z), "y": float(y), "x": float(x)}
            sample = prepare_sample(stem, int(tt), node, bundle, frame_cache, reader)
            if sample.shape != (5, *BIG):
                raise RuntimeError(f"unexpected enlarged sample shape {sample.shape}")
            out[i] = sample[:4]
            if checked < job["checks"]:
                ref = prepare_sample(stem, int(tt), node, ref_bundle, frame_cache, reader)
                if ref.shape != (5, *CROP) or not np.array_equal(sample[inner], ref):
                    raise RuntimeError("inner crop != DivNet-contract sample (contract drift)")
                checked += 1
            for key in [k for k in prepared if k[1] < int(tt) - 1]:
                prepared.pop(key)
        prepared.clear()
        frame_cache.clear()
        return {"stem": stem, "crops": out, "secs": time.time() - t0, "error": None, "checked": checked,
                "frames_read": reader.frames_read, "decoders": decoder_names()}
    except Exception as exc:
        return {"stem": stem, "crops": None, "secs": time.time() - t0, "error": f"{type(exc).__name__}: {exc}",
                "checked": 0, "frames_read": 0, "decoders": decoder_names()}


def step_build_cache(cfg: dict, state: dict) -> dict:
    import multiprocessing as mp

    t0 = time.time()
    meta = state["meta"]
    cache = Path(state["cache_dir"])
    crops_path = cache / "crops.npy"
    dtype = np.dtype(cfg["cache_dtype"])
    n = len(meta)
    need = n * 4 * int(np.prod(BIG)) * dtype.itemsize
    try:
        import shutil
        free = shutil.disk_usage(cache).free
        log(f"[build] cache disk free {free / 2**30:.1f} GiB, needed {need / 2**30:.2f} GiB")
        if free < 1.2 * need and dtype == np.float32:
            dtype = np.dtype(np.float16)
            log("[build] WARNING not enough disk for float32 -> caching float16 (OOF inputs rounded to fp16)")
    except Exception:
        pass
    state["cache_dtype"] = dtype.name
    log(f"[build] {n} samples -> {crops_path} ({n * 4 * int(np.prod(BIG)) * dtype.itemsize / 2**30:.2f} GiB {dtype})")
    crops = np.lib.format.open_memmap(crops_path, mode="w+", dtype=dtype, shape=(max(n, 1), 4, *BIG))
    jobs = []
    for stem, grp in meta.groupby("stem", sort=True):
        jobs.append({"stem": stem, "data_dir": state["data_dir"], "dtype": dtype.name, "checks": cfg["contract_checks"],
                     "index": grp.index.to_numpy(),
                     "nodes": list(zip(grp["t"].tolist(), grp["z"].tolist(), grp["y"].tolist(), grp["x"].tolist()))})
    n_workers = cfg["build_workers"] or min(4, os.cpu_count() or 1)
    n_workers = max(1, min(n_workers, len(jobs) or 1))
    per_worker_threads = max(1, (os.cpu_count() or 1) // n_workers)
    valid = np.zeros(n, bool)
    video_rows = []
    lite = [{k: v for k, v in j.items() if k != "index"} for j in jobs]
    index_of = {j["stem"]: j["index"] for j in jobs}
    done = 0
    ctx = mp.get_context("spawn")
    with ctx.Pool(n_workers, initializer=_build_worker_init, initargs=(per_worker_threads,)) as pool:
        for res in pool.imap_unordered(crop_video, lite):
            done += 1
            stem = res["stem"]
            idx = index_of[stem]
            grp = meta.loc[idx]
            counts = grp["kind"].value_counts().to_dict()
            if res["error"] is None:
                crops[idx] = res["crops"]
                valid[idx] = True
                status = "ok"
            else:
                status = "FAILED " + res["error"]
            video_rows.append({"stem": stem, "crop_secs": round(res["secs"], 2), "frames_read": res["frames_read"],
                               "contract_checks": res["checked"], "crop_error": res["error"], "n_samples": len(idx)})
            log(f"[build] {done}/{len(jobs)} {stem} n={len(idx)} "
                + " ".join(f"{k}={v}" for k, v in sorted(counts.items()))
                + f" frames={res['frames_read']} {res['secs']:.1f}s decoders={res['decoders']} {status}")
    crops.flush()
    del crops
    meta = meta.copy()
    meta["valid"] = valid
    meta.to_csv(cache / "meta.csv", index_label="row")
    import pandas as pd
    vid = pd.DataFrame(video_rows)
    stats = state["stats"].merge(vid, on="stem", how="left") if not vid.empty else state["stats"]
    stats.to_csv(Path(state["work_dir"]) / "labels_summary.csv", index=False)
    state["meta"] = meta
    state["stats"] = stats
    secs = time.time() - t0
    state["timing"]["build_s"] = round(secs, 1)
    ok = vid[vid["crop_error"].isna()] if not vid.empty else vid
    per_video = float(ok["crop_secs"].mean()) if len(ok) else float("nan")
    state["timing"]["crop_s_per_video_mean"] = round(per_video, 2)
    log(f"[build] done in {secs:.1f} s with {n_workers} workers; valid samples {int(valid.sum())}/{n}; "
        f"videos failed {int(vid['crop_error'].notna().sum()) if not vid.empty else 0}; mean crop time per video "
        f"{per_video:.2f} s (single worker)")
    return state


# ----------------------------------------------------------------------------------------------------------------------
# step 3: training (subprocess per fold)
# ----------------------------------------------------------------------------------------------------------------------
def _src_paths() -> list[str]:
    import tools.daughternet as dn  # noqa: F401  (the tools package the kernel materialised or the repo copy)
    paths = [str(Path(__file__).resolve().parent), str(Path(dn.__file__).resolve().parent.parent)]
    return list(dict.fromkeys(paths))


def _fold_dir(work: Path, fold: int) -> Path:
    return work / "folds" / f"fold_{fold}"


def _pump(stream, prefix: str, sink) -> None:
    for line in iter(stream.readline, ""):
        print(prefix + line.rstrip(), flush=True)
        sink.write(line)
        sink.flush()
    stream.close()


def step_train(cfg: dict, state: dict) -> dict:
    import torch

    t0 = time.time()
    work = Path(state["work_dir"])
    cache = Path(state["cache_dir"])
    meta = state["meta"]
    folds = sorted(int(f) for f in meta.loc[meta["valid"], "fold"].unique())
    (work / "folds").mkdir(parents=True, exist_ok=True)
    cfg_path = cache / "kdn_cfg.json"
    cfg_path.write_text(json.dumps({**cfg, "_cache_dir": str(cache), "_work_dir": str(work)}, indent=1))
    n_gpu = torch.cuda.device_count() if cfg["gpus"] != 0 else 0
    if cfg["gpus"] > 0:
        n_gpu = min(n_gpu, cfg["gpus"])
    slots = [str(i) for i in range(n_gpu)] or ["cpu"]
    log(f"[train] folds {folds} on slots {slots} (subprocess per fold)")
    env_base = os.environ.copy()
    env_base["PYTHONPATH"] = os.pathsep.join(_src_paths() + [p for p in env_base.get("PYTHONPATH", "").split(os.pathsep) if p])
    env_base["PYTHONUNBUFFERED"] = "1"
    queue = list(folds)
    running: dict[str, tuple] = {}
    results: dict[int, int] = {}
    while queue or running:
        for slot in slots:
            if slot not in running and queue:
                fold = queue.pop(0)
                fdir = _fold_dir(work, fold)
                fdir.mkdir(parents=True, exist_ok=True)
                # a rerun into the same work dir must never pick up a previous run's fold model / OOF rows
                # (a fold that fails or never saves best.pt would otherwise load or merge stale files silently)
                for stale in ("best.pt", "last.pt", "oof.csv", "summary.json", "history.json", "train.log"):
                    (fdir / stale).unlink(missing_ok=True)
                env = dict(env_base)
                if slot != "cpu":
                    env["CUDA_VISIBLE_DEVICES"] = slot
                device = "cuda:0" if slot != "cpu" else "cpu"
                cmd = [sys.executable, str(Path(__file__).resolve()), "worker", "--cfg", str(cfg_path),
                       "--fold", str(fold), "--device", device, "--threads",
                       str(max(1, (os.cpu_count() or 1) // len(slots)))]
                sink = open(fdir / "train.log", "a", encoding="utf-8")
                proc = subprocess.Popen(cmd, env=env, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True,
                                        encoding="utf-8", errors="replace")
                th = threading.Thread(target=_pump, args=(proc.stdout, f"[fold {fold}@{slot}] ", sink), daemon=True)
                th.start()
                running[slot] = (fold, proc, th, sink, time.time())
                log(f"[train] started fold {fold} on {slot} (pid {proc.pid})")
        time.sleep(1.0)
        for slot in list(running):
            fold, proc, th, sink, ts = running[slot]
            if proc.poll() is not None:
                th.join(timeout=10)
                sink.close()
                results[fold] = proc.returncode
                log(f"[train] fold {fold} finished rc={proc.returncode} in {time.time() - ts:.0f} s")
                del running[slot]
    state["train_rc"] = results
    state["timing"]["train_s"] = round(time.time() - t0, 1)
    failed = [f for f, rc in results.items() if rc != 0]
    if failed:
        log(f"[train] WARNING folds failed: {failed} (their videos get no OOF prediction)")
    return state


def _load_split(crops, rows: np.ndarray, inner_only: bool, dtype, device, torch, chunk: int = 1024):
    """Rows of the memmap -> one tensor on device; inner_only keeps the contract 16x32x32 crop."""
    if len(rows) == 0:
        shape = (0, 4, *CROP) if inner_only else (0, 4, *BIG)
        return torch.zeros(shape, dtype=dtype, device=device)
    parts = []
    for s in range(0, len(rows), chunk):
        block = np.asarray(crops[rows[s:s + chunk]], dtype=np.float32)
        if inner_only:
            block = block[:, :, MARGIN:MARGIN + CROP[0], MARGIN:MARGIN + CROP[1], MARGIN:MARGIN + CROP[2]]
        parts.append(torch.from_numpy(np.ascontiguousarray(block)).to(device=device, dtype=dtype))
    return torch.cat(parts)


def augment_batch(xb, rng: np.random.Generator, cfg: dict, marker, torch):
    """(B, 4, 18, 34, 34) image crops -> (B, 5, 16, 32, 32): +-jitter crop (image shifts, marker stays centred), then
    random flips on z/y/x and a random k*90 degree rotation in yx applied to all 5 channels, then a per-sample
    intensity gain on the 4 image channels (clipped back to the contract range [-0.5, 6])."""
    b = xb.shape[0]
    dev = xb.device
    j = int(cfg["jitter"])
    offs = torch.from_numpy(rng.integers(MARGIN - j, MARGIN + j + 1, size=(b, 3))).to(dev)
    zi = offs[:, 0, None] + torch.arange(CROP[0], device=dev)
    yi = offs[:, 1, None] + torch.arange(CROP[1], device=dev)
    xi = offs[:, 2, None] + torch.arange(CROP[2], device=dev)
    bi = torch.arange(b, device=dev)[:, None, None, None]
    img = xb[bi, :, zi[:, :, None, None], yi[:, None, :, None], xi[:, None, None, :]]  # (B, 16, 32, 32, 4)
    img = img.permute(0, 4, 1, 2, 3).float()
    x = torch.cat([img, marker.to(dev).expand(b, 1, *CROP)], dim=1)
    flips = torch.from_numpy(rng.random((b, 3)) < 0.5).to(dev)
    for axis, dim in enumerate((2, 3, 4)):
        m = flips[:, axis].view(b, 1, 1, 1, 1)
        x = torch.where(m, x.flip(dim), x)
    rot = rng.integers(0, 4, size=b)
    for k in (1, 2, 3):
        sel = np.flatnonzero(rot == k)
        if len(sel):
            idx = torch.from_numpy(sel).to(dev)
            x[idx] = torch.rot90(x[idx], k, dims=(3, 4))
    gain = torch.from_numpy(rng.uniform(cfg["gain"][0], cfg["gain"][1], size=(b, 1, 1, 1, 1)).astype(np.float32)).to(dev)
    x[:, :4] = (x[:, :4] * gain).clamp(-0.5, 6.0)
    return x


def contract_marker(torch):
    sigma = tuple(float(v) for v in DEFAULT_CFG["input"]["marker_sigma"])
    return torch.from_numpy(_marker(CROP, sigma)).view(1, 1, *CROP)


def predict_inner(model, x_inner, marker, batch: int, torch) -> np.ndarray:
    """Softmax (N, 3) for contract samples (image inner crops + centred marker), fp32 like score_nodes."""
    out = []
    model.eval()
    with torch.no_grad():
        for s in range(0, x_inner.shape[0], batch):
            xb = x_inner[s:s + batch].float()
            x = torch.cat([xb, marker.to(xb.device).expand(xb.shape[0], 1, *CROP)], dim=1)
            out.append(torch.softmax(model(x).float(), dim=1).cpu().numpy())
    return np.concatenate(out) if out else np.zeros((0, len(CLASSES)), np.float32)


def auc_score(scores, positive) -> float | None:
    import pandas as pd
    s = np.asarray(scores, dtype=np.float64)
    y = np.asarray(positive, dtype=bool)
    n1, n0 = int(y.sum()), int((~y).sum())
    if n1 == 0 or n0 == 0:
        return None
    ranks = pd.Series(s).rank(method="average").to_numpy()
    return float((ranks[y].sum() - n1 * (n1 + 1) / 2.0) / (n1 * n0))


def _val_metrics(probs: np.ndarray, y: np.ndarray) -> dict:
    if len(y) == 0:
        return {"n": 0}
    p_true = np.clip(probs[np.arange(len(y)), y], 1e-7, 1.0)
    ce = -np.log(p_true)
    present = [c for c in range(len(CLASSES)) if (y == c).any()]
    macro = float(np.mean([ce[y == c].mean() for c in present]))
    res = {"n": int(len(y)), "macro_ce": macro, "ce": float(ce.mean()),
           "acc": float((probs.argmax(1) == y).mean())}
    for c, name in enumerate(CLASSES):
        res[f"n_{name}"] = int((y == c).sum())
        res[f"ce_{name}"] = float(ce[y == c].mean()) if (y == c).any() else None
        res[f"auc_{name}"] = auc_score(probs[:, c], y == c)
    return res


def _atomic_save(obj, path: Path, torch) -> None:
    tmp = path.with_suffix(path.suffix + ".tmp")
    torch.save(obj, tmp)
    os.replace(tmp, path)


def train_fold(cfg: dict, fold: int, device_str: str, threads: int = 0) -> dict:
    import pandas as pd
    import torch

    t_start = time.time()
    if threads:
        torch.set_num_threads(int(threads))
    cache = Path(cfg["_cache_dir"])
    work = Path(cfg["_work_dir"])
    fdir = _fold_dir(work, fold)
    fdir.mkdir(parents=True, exist_ok=True)
    device = torch.device(device_str)
    use_cuda = device.type == "cuda"
    if use_cuda:
        torch.backends.cudnn.benchmark = True
    seed = int(cfg["seed"]) * 1000 + int(fold)
    torch.manual_seed(seed)
    rng = np.random.default_rng(seed)
    meta = pd.read_csv(cache / "meta.csv", index_col="row")
    crops = np.load(cache / "crops.npy", mmap_mode="r")
    meta = meta[meta["valid"].astype(bool)]
    n_folds = int(cfg["n_folds"])
    if cfg.get("inner_val") and n_folds < 3:
        log(f"fold {fold}: KDN_INNER_VAL needs >= 3 folds (have {n_folds}) -> early stopping on the OOF fold")
        cfg = {**cfg, "inner_val": False}
    if cfg.get("inner_val"):
        es_fold = (fold + 1) % n_folds
        train_mask = ~meta["fold"].isin([fold, es_fold])
        es_mask = meta["fold"] == es_fold
    else:
        es_fold = fold
        train_mask = meta["fold"] != fold
        es_mask = meta["fold"] == fold
    oof_mask = meta["fold"] == fold
    tr_rows = meta.index[train_mask].to_numpy()
    es_rows = meta.index[es_mask].to_numpy()
    oof_rows = meta.index[oof_mask].to_numpy()
    y_tr = meta.loc[tr_rows, "label_idx"].to_numpy().astype(np.int64)
    y_es = meta.loc[es_rows, "label_idx"].to_numpy().astype(np.int64)
    log(f"fold {fold}: device {device} train {len(tr_rows)} {np.bincount(y_tr, minlength=3).tolist()} | "
        f"early-stop fold {es_fold}: {len(es_rows)} {np.bincount(y_es, minlength=3).tolist()} | oof {len(oof_rows)}")
    if len(tr_rows) == 0:
        raise RuntimeError("no training samples")
    store_dtype = torch.float16 if use_cuda else torch.float32
    t_load = time.time()
    x_tr = _load_split(crops, tr_rows, False, store_dtype, device, torch)
    x_es = _load_split(crops, es_rows, True, torch.float32, device, torch)
    y_tr_t = torch.from_numpy(y_tr).to(device)
    log(f"fold {fold}: loaded crops in {time.time() - t_load:.1f} s (train {tuple(x_tr.shape)} {x_tr.dtype})")
    marker = contract_marker(torch).to(device)

    ckpt_cfg = copy.deepcopy(DEFAULT_CFG)
    ckpt_cfg["classes"] = list(CLASSES)
    ckpt_cfg["train"] = {k: cfg[k] for k in ("epochs", "patience", "batch", "lr", "weight_decay", "warmup_frac",
                                               "epoch_samples", "jitter", "gain", "seed", "cont_per_video",
                                               "hardneg_per_video", "cont_min_dist", "inner_val", "n_folds")}
    ckpt_cfg["train"].update({"fold": fold, "early_stop_fold": es_fold, "kernel": "biohub-l2-kdn-train"})
    model = build_model(ckpt_cfg).to(device)
    opt = torch.optim.AdamW(model.parameters(), lr=cfg["lr"], weight_decay=cfg["weight_decay"])
    classes_present = [c for c in range(len(CLASSES)) if (y_tr == c).any()]
    pools = {c: np.flatnonzero(y_tr == c) for c in classes_present}
    epoch_samples = int(cfg["epoch_samples"]) or len(tr_rows)
    steps_per_epoch = max(1, math.ceil(epoch_samples / cfg["batch"]))
    total_steps = steps_per_epoch * int(cfg["epochs"])
    warmup = max(1, int(cfg["warmup_frac"] * total_steps))

    def lr_at(step: int) -> float:
        if step < warmup:
            return cfg["lr"] * (step + 1) / warmup
        prog = (step - warmup) / max(1, total_steps - warmup)
        return cfg["lr"] * 0.5 * (1.0 + math.cos(math.pi * min(1.0, prog)))

    use_amp = bool(cfg["amp"]) and use_cuda
    try:
        scaler = torch.amp.GradScaler("cuda", enabled=use_amp)
    except Exception:  # older torch
        scaler = torch.cuda.amp.GradScaler(enabled=use_amp)
    best, best_epoch, bad, step = float("inf"), -1, 0, 0
    history = []
    for epoch in range(int(cfg["epochs"])):
        te = time.time()
        model.train()
        cls = rng.choice(classes_present, size=epoch_samples)
        order = np.empty(epoch_samples, dtype=np.int64)
        for c in classes_present:
            sel = np.flatnonzero(cls == c)
            order[sel] = rng.choice(pools[c], size=len(sel), replace=True)
        rng.shuffle(order)
        run_loss, seen = torch.zeros((), device=device), 0
        for s in range(0, epoch_samples, cfg["batch"]):
            idx = order[s:s + cfg["batch"]]
            idx_t = torch.from_numpy(idx).to(device)
            xb = augment_batch(x_tr[idx_t], rng, cfg, marker, torch)
            yb = y_tr_t[idx_t]
            for group in opt.param_groups:
                group["lr"] = lr_at(step)
            opt.zero_grad(set_to_none=True)
            with torch.autocast(device_type=device.type, dtype=torch.float16, enabled=use_amp):
                logits = model(xb)
            loss = torch.nn.functional.cross_entropy(logits.float(), yb)
            scaler.scale(loss).backward()
            scaler.step(opt)
            scaler.update()
            run_loss += loss.detach() * len(idx)
            seen += len(idx)
            step += 1
        train_loss = float(run_loss) / max(1, seen)
        if use_cuda:
            torch.cuda.synchronize()
        train_secs = time.time() - te
        probs = predict_inner(model, x_es, marker, cfg["eval_batch"], torch)
        vm = _val_metrics(probs, y_es)
        val_loss = vm.get("macro_ce", float("nan")) if vm["n"] else float("nan")
        improved = vm["n"] > 0 and val_loss < best
        state_cpu = {k: v.detach().cpu() for k, v in model.state_dict().items()}
        payload = {"model_state": state_cpu, "config": ckpt_cfg, "fold": int(fold), "epoch": int(epoch),
                   "val_loss": float(val_loss)}
        _atomic_save(payload, fdir / "last.pt", torch)
        if improved or (vm["n"] == 0):
            best, best_epoch, bad = (val_loss if vm["n"] else best), epoch, 0
            _atomic_save(payload, fdir / "best.pt", torch)
        else:
            bad += 1
        rec = {"epoch": epoch, "train_loss": train_loss, "val_loss": val_loss, "lr": lr_at(step - 1),
               "secs": round(time.time() - te, 1), "train_secs": round(train_secs, 1),
               "samples_per_s": round(seen / max(1e-9, train_secs), 1),
               **{f"val_{k}": v for k, v in vm.items()}}
        history.append(rec)
        (fdir / "history.json").write_text(json.dumps(history, indent=1))
        auc_txt = " ".join(f"auc_{n}={vm.get('auc_' + n) if vm.get('auc_' + n) is None else round(vm['auc_' + n], 3)}"
                           for n in CLASSES)
        log(f"fold {fold} epoch {epoch}: train {train_loss:.4f} val(macro CE) {val_loss:.4f} "
            f"{'*' if improved else ' '} {auc_txt} {rec['secs']}s (train {rec['samples_per_s']} samples/s)")
        if vm["n"] and bad >= int(cfg["patience"]):
            log(f"fold {fold}: early stop at epoch {epoch} (best {best_epoch}, val {best:.4f})")
            break
    del x_tr
    if not (fdir / "best.pt").is_file():
        raise RuntimeError(f"fold {fold}: no best.pt written in this run (epochs={cfg['epochs']}, history "
                           f"{[h['val_loss'] for h in history]}) -> no OOF for this fold")
    # OOF prediction with best.pt through the shared loader (same path the submission patch uses)
    bundle = load_daughternet([fdir / "best.pt"], device=str(device))
    x_oof = _load_split(crops, oof_rows, True, torch.float32, device, torch)
    p_oof = predict_inner(bundle["models"][0], x_oof, marker, cfg["eval_batch"], torch)
    oof = pd.DataFrame({"row": oof_rows, "p_cont": p_oof[:, 0], "p_par": p_oof[:, 1], "p_dau": p_oof[:, 2]})
    oof.to_csv(fdir / "oof.csv", index=False)
    y_oof = meta.loc[oof_rows, "label_idx"].to_numpy().astype(np.int64)
    om = _val_metrics(p_oof, y_oof)
    summary = {"fold": fold, "best_epoch": best_epoch, "best_val_loss": best, "epochs_run": len(history),
               "early_stop_fold": es_fold, "n_train": int(len(tr_rows)), "n_oof": int(len(oof_rows)),
               "train_counts": np.bincount(y_tr, minlength=3).tolist(), "oof_metrics": om,
               "secs": round(time.time() - t_start, 1), "device": str(device)}
    (fdir / "summary.json").write_text(json.dumps(summary, indent=1))
    log(f"fold {fold}: OOF {len(oof_rows)} samples macro CE {om.get('macro_ce')} "
        f"auc_parent {om.get('auc_parent')} auc_daughter {om.get('auc_daughter')}; total {summary['secs']} s")
    return summary


def worker_main(argv: list[str]) -> int:
    import argparse

    ap = argparse.ArgumentParser()
    ap.add_argument("--cfg", required=True)
    ap.add_argument("--fold", type=int, required=True)
    ap.add_argument("--device", default="cpu")
    ap.add_argument("--threads", type=int, default=0)
    a = ap.parse_args(argv)
    cfg = json.loads(Path(a.cfg).read_text())
    train_fold(cfg, a.fold, a.device, a.threads)
    return 0


# ----------------------------------------------------------------------------------------------------------------------
# step 4: OOF merge, readout, DivNet comparison
# ----------------------------------------------------------------------------------------------------------------------
TOP_FRACS = (("top0.1%", 0.001), ("top1%", 0.01), ("top5%", 0.05))


def _rank_block(scores: np.ndarray, positive: np.ndarray) -> dict:
    n = len(scores)
    n_pos = int(positive.sum())
    base = n_pos / n if n else None
    out = {"n": int(n), "n_pos": n_pos, "base_rate": base, "auc": auc_score(scores, positive)}
    order = np.argsort(-scores, kind="stable")
    for name, frac in TOP_FRACS:
        if n == 0:
            out[name] = None
            continue
        k = max(1, int(math.ceil(frac * n)))
        tp = int(positive[order[:k]].sum())
        fp = k - tp
        prec = tp / k
        tpr = tp / n_pos if n_pos else None
        fpr = fp / (n - n_pos) if n - n_pos else None
        lr_plus = (tpr / fpr) if (tpr is not None and fpr) else None
        out[name] = {"k": k, "tp": tp, "precision": prec, "lift": (prec / base) if base else None,
                     "lr_plus": lr_plus, "lr_plus_infinite": bool(fpr == 0 and tp > 0)}
    return out


def _find_divnet_checkpoint(cfg: dict) -> Path | None:
    if cfg.get("divnet_checkpoint"):
        p = Path(cfg["divnet_checkpoint"])
        return p if p.is_file() else None
    for p in _checkpoint_candidates():
        if Path(p).is_file():
            return Path(p)
    root = Path("/kaggle/input")
    if root.exists():  # pruned walk: a plain rglob would descend into every competition .zarr chunk tree
        found = []
        for dirpath, dirnames, files in os.walk(root, followlinks=True):
            depth = Path(dirpath).relative_to(root).parts
            dirnames[:] = [] if len(depth) >= 6 else sorted(
                d for d in dirnames if not d.endswith((".zarr", ".geff")) and not d.startswith("."))
            if "best_overall.pt" in files and "divnet" in dirpath.lower():
                found.append(Path(dirpath) / "best_overall.pt")
        if found:
            return sorted(found)[0]
    return None


def step_oof(cfg: dict, state: dict) -> dict:
    import pandas as pd
    import torch

    t0 = time.time()
    work = Path(state["work_dir"])
    cache = Path(state["cache_dir"])
    meta = pd.read_csv(cache / "meta.csv", index_col="row")
    meta = meta[meta["valid"].astype(bool)].copy()
    for col in ("p_cont", "p_par", "p_dau"):
        meta[col] = np.nan
    fold_summaries = {}
    for fold in sorted(meta["fold"].unique()):
        fdir = _fold_dir(work, int(fold))
        f_oof = fdir / "oof.csv"
        if f_oof.is_file():
            o = pd.read_csv(f_oof).set_index("row")
            meta.loc[o.index, ["p_cont", "p_par", "p_dau"]] = o[["p_cont", "p_par", "p_dau"]].to_numpy()
        if (fdir / "summary.json").is_file():
            fold_summaries[int(fold)] = json.loads((fdir / "summary.json").read_text())
    has_p = meta["p_par"].notna()
    log(f"[oof] predictions for {int(has_p.sum())}/{len(meta)} samples")

    divnet_info = {"enabled": bool(cfg["divnet_compare"])}
    if cfg["divnet_compare"]:
        try:
            ckpt = _find_divnet_checkpoint(cfg)
            if ckpt is None:
                raise FileNotFoundError("best_overall.pt not found")
            bundle = load_divnet_mitosis_model(ckpt)
            dev = bundle["device"]
            crops = np.load(cache / "crops.npy", mmap_mode="r")
            marker = contract_marker(torch).to(dev)
            rows = meta.index.to_numpy()
            ps = []
            for s in range(0, len(rows), 1024):
                x_in = _load_split(crops, rows[s:s + 1024], True, torch.float32, dev, torch)
                with torch.no_grad():
                    for b in range(0, x_in.shape[0], cfg["eval_batch"]):
                        xb = x_in[b:b + cfg["eval_batch"]]
                        x = torch.cat([xb, marker.expand(xb.shape[0], 1, *CROP)], dim=1)
                        ps.append(torch.sigmoid(bundle["model"](x)).flatten().float().cpu().numpy())
            meta["p_divnet"] = np.concatenate(ps) if ps else np.zeros(0)
            del crops, bundle
            divnet_info.update({"checkpoint": str(ckpt), "in_sample": True,
                                "note": "public giorgosi DivNet best_overall.pt was trained on these 199 train videos "
                                        "(4 folds, best fold model): its numbers are IN-SAMPLE, DaughterNet's are OOF"})
            log(f"[oof] DivNet scored {len(meta)} samples with {ckpt}")
        except Exception as exc:
            divnet_info["error"] = f"{type(exc).__name__}: {exc}"
            log(f"[oof] DivNet comparison skipped: {divnet_info['error']}")

    out = meta.reset_index(drop=True)
    extra = [c for c in ("node_id", "kind", "embryo", "label_idx", "p_divnet") if c in out.columns]
    out[OOF_COLUMNS + extra].to_csv(work / "oof.csv", index=False)

    readout = {
        "kernel": "biohub-l2-kdn-train", "classes": list(CLASSES), "n_samples": int(len(out)),
        "n_with_oof": int(out["p_par"].notna().sum()), "splits_source": state.get("splits_source"),
        "definitions": {
            "base_rate": "fraction of the embryo's labelled samples that carry the class (sampling-dependent: cont is "
                         "subsampled to <= cont_per_video + hardneg_per_video per video)",
            "precision": "fraction of the class among the top-k samples ranked by that class probability",
            "lift": "precision / base_rate",
            "lr_plus": "likelihood ratio at the cutoff = TPR / FPR = (tp/P) / (fp/N); null when fp = 0",
            "auc": "one-vs-rest ROC AUC of the class probability",
        },
        "counts": {e: g.groupby("kind").size().to_dict() for e, g in out.groupby("embryo")},
        "per_embryo": {}, "hardneg": {}, "folds": fold_summaries, "divnet_comparison": divnet_info,
        "timing": state.get("timing", {}), "train_rc": state.get("train_rc"),
        "config": {k: v for k, v in cfg.items() if not k.startswith("_")},
    }
    scored = out[out["p_par"].notna()]
    groups = [(e, g) for e, g in scored.groupby("embryo")] + [("all", scored)]
    prob_col = {"cont": "p_cont", "parent": "p_par", "daughter": "p_dau"}
    for emb, g in groups:
        readout["per_embryo"][emb] = {
            name: _rank_block(g[prob_col[name]].to_numpy(), (g["label"] == name).to_numpy()) for name in CLASSES}
        hu = g[g["kind"].str.startswith("hardneg") & g["kind"].str.contains("u")]
        hv = g[g["kind"].str.startswith("hardneg") & g["kind"].str.contains("v")]
        par, dau = g[g["label"] == "parent"], g[g["label"] == "daughter"]
        readout["hardneg"][emb] = {
            "n_hardneg_u": int(len(hu)), "n_hardneg_v": int(len(hv)),
            "auc_p_par_parent_vs_hardneg_u": auc_score(np.r_[par["p_par"], hu["p_par"]],
                                                       np.r_[np.ones(len(par)), np.zeros(len(hu))]),
            "auc_p_dau_daughter_vs_hardneg_v": auc_score(np.r_[dau["p_dau"], hv["p_dau"]],
                                                         np.r_[np.ones(len(dau)), np.zeros(len(hv))]),
        }
        if "p_divnet" in g.columns and divnet_info.get("in_sample"):
            sub = g[g["label"].isin(["parent", "cont"])]
            sub_rand = g[g["kind"].isin(["parent", "cont"])]
            y = (sub["label"] == "parent").to_numpy()
            y_r = (sub_rand["kind"] == "parent").to_numpy()
            divnet_info.setdefault("per_embryo", {})[emb] = {
                "subset": "parent vs cont (random cont + hard negatives)",
                "divnet_in_sample": _rank_block(sub["p_divnet"].to_numpy(), y),
                "daughternet_oof_p_par": _rank_block(sub["p_par"].to_numpy(), y),
                "random_cont_only": {"divnet_in_sample_auc": auc_score(sub_rand["p_divnet"].to_numpy(), y_r),
                                     "daughternet_oof_auc": auc_score(sub_rand["p_par"].to_numpy(), y_r)},
            }
    state["timing"]["oof_s"] = round(time.time() - t0, 1)
    readout["timing"] = state["timing"]
    (work / "oof_readout.json").write_text(json.dumps(readout, indent=1, default=_json_default))
    log(f"[oof] wrote {work / 'oof.csv'} ({len(out)} rows) and {work / 'oof_readout.json'}")
    _print_readout(readout)
    if not cfg.get("keep_cache"):
        import shutil
        shutil.rmtree(cache, ignore_errors=True)
    state["readout"] = readout
    return state


def _json_default(value):
    if isinstance(value, np.integer):
        return int(value)
    if isinstance(value, np.floating):
        return float(value)
    if isinstance(value, np.ndarray):
        return value.tolist()
    if isinstance(value, Path):
        return str(value)
    return str(value)


def _print_readout(readout: dict) -> None:
    for emb, block in readout["per_embryo"].items():
        for name, r in block.items():
            tops = " ".join(f"{k}:{r[k]['tp']}/{r[k]['k']}" for k, _ in TOP_FRACS if r.get(k))
            auc = None if r["auc"] is None else round(r["auc"], 4)
            base = "n/a" if r["base_rate"] is None else f"{r['base_rate']:.4f}"
            log(f"[readout] {emb:5s} {name:8s} n_pos={r['n_pos']:5d} base={base} AUC={auc} {tops}")
    for emb, r in readout.get("divnet_comparison", {}).get("per_embryo", {}).items():
        a, b = r["divnet_in_sample"]["auc"], r["daughternet_oof_p_par"]["auc"]
        log(f"[readout] {emb:5s} parent-vs-cont AUC DivNet(in-sample)={a} DaughterNet(OOF)={b}")


def step_verify(cfg: dict, state: dict, per_fold: int = 2) -> dict:
    """End-to-end contract check on a few OOF samples per fold: raw frames -> tools.daughternet.score_nodes with the
    fold's best.pt (the submission-patch path) must reproduce oof.csv, and DivNet's divnet_score_division must
    reproduce p_divnet. Differences are only fp rounding (float32 cache) or fp16 rounding (float16 cache)."""
    import pandas as pd
    import torch
    from tools.divnet_infer import divnet_score_division

    work = Path(state["work_dir"])
    oof = pd.read_csv(work / "oof.csv")
    oof = oof[oof["p_par"].notna()]
    reader = FrameReader(Path(state["data_dir"]))
    report = {"checked": 0, "max_abs_diff_daughternet": 0.0, "max_abs_diff_divnet": None, "rows": []}
    divnet_bundle = None
    if "p_divnet" in oof.columns and oof["p_divnet"].notna().any():
        ckpt = _find_divnet_checkpoint(cfg)
        divnet_bundle = load_divnet_mitosis_model(ckpt) if ckpt else None
        report["max_abs_diff_divnet"] = 0.0
    for fold, g in oof.groupby("fold"):
        best = _fold_dir(work, int(fold)) / "best.pt"
        if not best.is_file():
            continue
        bundle = load_daughternet([best], device="cpu")
        # prefer parents/daughters so the check covers the classes that matter
        pick = pd.concat([g[g["label"] != "cont"].head(per_fold), g[g["label"] == "cont"].head(1)])
        for _, r in pick.iterrows():
            node = {"t": int(r["t"]), "z": float(r["z"]), "y": float(r["y"]), "x": float(r["x"])}
            bundle["prepared_frames"] = {}
            p = score_nodes(r["stem"], [node], bundle, {}, reader, batch=1)[0]
            diff = float(np.abs(p - r[["p_cont", "p_par", "p_dau"]].to_numpy(dtype=np.float64)).max())
            report["max_abs_diff_daughternet"] = max(report["max_abs_diff_daughternet"], diff)
            row = {"fold": int(fold), "stem": r["stem"], "t": int(r["t"]), "label": r["label"], "diff": diff}
            if divnet_bundle is not None:
                divnet_bundle["prepared_frames"] = {}
                q = divnet_score_division(r["stem"], int(r["t"]), node, None, None, divnet_bundle, {}, reader)
                d2 = abs(q - float(r["p_divnet"]))
                report["max_abs_diff_divnet"] = max(report["max_abs_diff_divnet"], d2)
                row["diff_divnet"] = d2
            report["rows"].append(row)
            report["checked"] += 1
    # OOF p come from the fold worker (cuDNN on GPU, batch 256); the re-score runs on CPU one node at a time, so
    # only fp rounding separates them; a contract drift (crop shift, normalisation) moves p far more than this.
    tol = 1e-3 if state.get("cache_dtype", cfg["cache_dtype"]) == "float32" else 2e-2
    report["tolerance"] = tol
    report["pass"] = bool(report["checked"] > 0 and report["max_abs_diff_daughternet"] <= tol
                          and (report["max_abs_diff_divnet"] is None or report["max_abs_diff_divnet"] <= tol))
    (work / "contract_check.json").write_text(json.dumps(report, indent=1, default=_json_default))
    log(f"[verify] {report['checked']} samples re-scored from raw frames via score_nodes: max |dp| DaughterNet "
        f"{report['max_abs_diff_daughternet']:.2e}, DivNet {report['max_abs_diff_divnet']} -> "
        f"{'PASS' if report['pass'] else 'FAIL'} (tol {tol})")
    state["verify"] = report
    return state


def run_all(cfg: dict) -> dict:
    state = step_prepare(cfg)
    state = step_build_cache(cfg, state)
    state = step_train(cfg, state)
    state = step_oof(cfg, state)
    state = step_verify(cfg, state)
    return state


if __name__ == "__main__":
    if len(sys.argv) > 1 and sys.argv[1] == "worker":
        sys.exit(worker_main(sys.argv[2:]))
    run_all(config_from_env())
```

```python
%%writefile /kaggle/working/kds_src/kds_core.py
"""K-DS: score every exported L2 candidate node with the K-DN DaughterNet fold models (OOF or 5-fold ensemble).

One module shared by the Kaggle kernel kdn-score.ipynb (materialised with %%writefile under /kaggle/working/kds_src next
to kdn_core.py, tools/divnet_infer.py, tools/daughternet.py and splits.json) and by the laptop test test_local.py.

Modes (env KDS_ENSEMBLE)
  0    (default) oof   every row of coords_exp39.parquet is scored by the model of its video's fold, i.e. the fold model
                       that never trained on that video                        -> node_p_oof.parquet
  1              ens   softmax averaged over all fold models (what the submission kernel does) -> node_p_ens.parquet
  both           one joint pass: every fold model runs once per sample; oof = the video's fold model's softmax, ens
                 = their mean (the frames are read and prepared once for both outputs)
Output: columns stem, node_id, fold, p_cont, p_par, p_dau (softmax in tools.daughternet.CLASSES order), one row per
selected input row IN INPUT ORDER (all 1,012,255 rows unless a smoke/debug selection is set), NaN where a video has no
model or failed, plus node_p_<mode>_summary.json (rows per fold, NaN count, timing, checks, verification).
In ens mode `fold` is still the video's split fold (for reference); p averages all fold models, including the ones that
trained on that video.

Pipeline (run)
  discover  fold_k/best.pt under /kaggle/input (pruned walk that never enters *.zarr / *.geff; only best.pt files whose
            path has a fold_k directory), the coords parquet (dataset vitaliineibauer/biohub-l2-coords), the trainer's
            splits_used.json (falls back to the embedded experiments/l2_daughternet/splits.json per video); the same
            file seen under two mounts (/kaggle/input/<slug>/ and /kaggle/input/notebooks/<owner>/<slug>/) counts once;
            fold models must come from one trainer run whose n_folds matches the routing splits
  audit     OOF honesty, independent of the routing: a video whose fold model trained on it or on a video of its leak
            group (splits.json meta.leak_groups) gets no OOF prediction (NaN, listed in the summary)
  plan      videos -> worker processes (one per GPU, CUDA_VISIBLE_DEVICES 0 / 1), greedy on rows x models + frames;
            a video is scheduled for the modes whose part is missing (a partial rerun redoes only what is missing)
  worker    per video: each needed frame is read once (kdn_core.FrameReader) and prepared exactly like the DivNet contract
            (_pool_xy + _normalise with the checkpoint config) by a thread pool that also prefetches the next video;
            samples are cropped from the prepared frames with the contract's own _crop_with_padding and _marker; batches
            of KDS_BATCH go to the GPU (fp32 unless KDS_AMP=1) and the CPU crop of batch k+1 overlaps the GPU pass of
            batch k; the result is written to parts/<mode>/<stem>.parquet (atomic; a rerun skips valid parts, parts
            are wiped when the checkpoints / coords / folds / amp signature changes)
  merge     parts -> node_p_<mode>.parquet aligned to the input rows on (stem, node_id)
  verify    a few rows per fold re-scored from raw frames through tools.daughternet.score_nodes on CPU -> max |dp|
Contract guard: in every video KDS_CHECKS_FRESH samples are rebuilt by tools.daughternet.prepare_sample from raw frames
and KDS_CHECKS_CACHED from the shared prepared frames; any difference from the fast-path sample aborts the worker.
"""
from __future__ import annotations

import hashlib
import json
import os
import re
import subprocess
import sys
import threading
import time
import zlib
from concurrent.futures import ThreadPoolExecutor
from pathlib import Path

import numpy as np


def _bootstrap_paths() -> list[str]:
    """Make tools/ (divnet_infer, daughternet) and kdn_core importable in the repo layout and in the kernel layout."""
    here = Path(__file__).resolve().parent
    paths = [here]
    for base in (here, *here.parents):
        if (base / "tools" / "daughternet.py").is_file():
            paths.append(base)
            break
    for cand in (here / "kdn_core.py", here.parent / "kdn_train" / "kdn_core.py"):
        if cand.is_file():
            paths.append(cand.parent)
            break
    out = list(dict.fromkeys(str(p) for p in paths))
    for p in reversed(out):
        if p not in sys.path:
            sys.path.insert(0, p)
    return out


SRC_PATHS = _bootstrap_paths()

import kdn_core  # noqa: E402
from kdn_core import FrameReader, decoder_names, load_splits, log, resolve_data_dir  # noqa: E402
from tools.daughternet import CLASSES, load_daughternet, prepare_sample, score_nodes  # noqa: E402
from tools.divnet_infer import _crop_with_padding, _marker, _normalise, _pool_xy  # noqa: E402

KERNEL = "biohub-l2-kdn-score"
P_COLUMNS = ["p_cont", "p_par", "p_dau"]
OUT_COLUMNS = ["stem", "node_id", "fold", *P_COLUMNS]
OUTPUTS = {"oof": "node_p_oof.parquet", "ens": "node_p_ens.parquet"}
PRUNE_SUFFIXES = (".zarr", ".geff")
FOLD_DIR_RE = re.compile(r"fold_(\d+)")
FRAME_COST_ROWS = 100      # planning weight: one prepared frame ~ the GPU time of 100 samples
if tuple(CLASSES) != ("cont", "parent", "daughter"):
    raise RuntimeError(f"tools.daughternet.CLASSES changed to {CLASSES}: update P_COLUMNS")


class ContractError(RuntimeError):
    """Fast-path sample differs from tools.daughternet.prepare_sample: never write predictions from it."""


# ----------------------------------------------------------------------------------------------------------------------
# configuration
# ----------------------------------------------------------------------------------------------------------------------
def _env_flag(name: str, default: str = "0") -> bool:
    return os.environ.get(name, default).strip() not in ("", "0", "false", "False", "no")


def parse_modes(value: str) -> list[str]:
    v = str(value).strip().lower()
    if v in ("", "0", "false", "no", "oof"):
        return ["oof"]
    if v in ("1", "true", "yes", "ens", "ensemble"):
        return ["ens"]
    if v in ("both", "2", "oof+ens", "all"):
        return ["oof", "ens"]
    raise ValueError(f"KDS_ENSEMBLE={value!r}: use 0 (oof), 1 (ensemble) or both")


def config_from_env(overrides: dict | None = None) -> dict:
    env = os.environ.get
    smoke = _env_flag("KDS_SMOKE")
    on_kaggle = Path("/kaggle/working").exists()
    cfg = {
        "smoke": smoke,
        "modes": parse_modes(env("KDS_ENSEMBLE", "0")),
        "data_dir": env("KDS_DATA_DIR", ""),                   # competition train dir (default: kdn_core rule)
        "work_dir": env("KDS_WORK_DIR", "/kaggle/working" if on_kaggle else str(Path.cwd() / "kds_out")),
        "input_root": env("KDS_INPUT_ROOT", "/kaggle/input"),
        "ckpt_root": env("KDS_CKPT_ROOT", ""),                 # where to look for fold_k/best.pt (default input_root)
        "ckpt_name": env("KDS_CKPT_NAME", "best.pt"),
        "train_slug": env("KDS_TRAIN_SLUG", "biohub-l2-kdn-train"),
        "coords": env("KDS_COORDS", ""),                       # explicit parquet path
        "coords_name": env("KDS_COORDS_NAME", "coords_exp39.parquet"),
        "coords_slug": env("KDS_COORDS_SLUG", "biohub-l2-coords"),
        "splits": env("KDS_SPLITS", ""),                       # explicit splits file (else trainer splits_used.json)
        "stems": [s.strip() for s in env("KDS_STEMS", "").split(",") if s.strip()],
        "max_videos": int(env("KDS_MAX_VIDEOS", "3" if smoke else "0") or 0),
        "rows_per_video": int(env("KDS_ROWS_PER_VIDEO", "300" if smoke else "0") or 0),
        "seed": int(env("KDS_SEED", "0")),
        "batch": int(env("KDS_BATCH", "512")),
        "pad_to": int(env("KDS_PAD_TO", "64")),                # last batch of a video padded -> few cuDNN shapes
        "amp": _env_flag("KDS_AMP", "0"),                      # fp16 autocast (faster, |dp| ~1e-3 vs score_nodes)
        "gpus": int(env("KDS_GPUS", "-1")),                    # -1 = all visible, 0 = CPU
        "workers": int(env("KDS_WORKERS", "0")),               # 0 = one per GPU (max 2), 1 on CPU
        "max_gpu_workers": int(env("KDS_MAX_GPU_WORKERS", "2")),
        "frame_threads": int(env("KDS_FRAME_THREADS", "2")),   # per worker; 0 = prepare frames inline
        "checks_fresh": int(env("KDS_CHECKS_FRESH", "1")),     # per video, contract sample rebuilt from raw frames
        "checks_cached": int(env("KDS_CHECKS_CACHED", "2")),   # per video, contract sample from the shared frames
        "verify_per_fold": int(env("KDS_VERIFY_PER_FOLD", "3")),
        "verify_tol": float(env("KDS_VERIFY_TOL", "0") or 0),  # 0 = auto: 1e-4 fp32 (GPU vs CPU), 2e-2 amp
        "expect_folds": int(env("KDS_EXPECT_FOLDS", "5")),
        "retries": int(env("KDS_RETRIES", "1")),
        "resume": _env_flag("KDS_RESUME", "1"),
    }
    if overrides:
        cfg.update(overrides)
    if isinstance(cfg.get("modes"), str):
        cfg["modes"] = parse_modes(cfg["modes"])
    return cfg


# ----------------------------------------------------------------------------------------------------------------------
# discovery
# ----------------------------------------------------------------------------------------------------------------------
def pruned_walk(root: Path, max_depth: int = 8):
    """os.walk that never descends into *.zarr / *.geff (the competition chunk trees), hidden dirs or __pycache__."""
    root = Path(root)
    if not root.exists():
        return
    for dirpath, dirnames, filenames in os.walk(root, followlinks=True):
        depth = len(Path(dirpath).relative_to(root).parts)
        if depth >= max_depth:
            dirnames[:] = []
        else:
            dirnames[:] = sorted(d for d in dirnames if not d.endswith(PRUNE_SUFFIXES) and not d.startswith(".")
                                 and d != "__pycache__")
        yield Path(dirpath), filenames


def find_files(root: Path, name: str, max_depth: int = 8) -> list[Path]:
    return [d / name for d, files in pruned_walk(root, max_depth) if name in files]


def _file_sha(path: Path) -> str:
    h = hashlib.sha256()
    with open(path, "rb") as fh:
        for block in iter(lambda: fh.read(1 << 20), b""):
            h.update(block)
    return h.hexdigest()


def code_shas() -> dict[str, str]:
    """sha256 of the code that determines the scores (part signature: a code change never reuses stale parts)."""
    import tools.daughternet as _dn
    import tools.divnet_infer as _dv

    out = {}
    for name, f in (("kds_core.py", __file__), ("daughternet.py", _dn.__file__), ("divnet_infer.py", _dv.__file__),
                    ("kdn_core.py", kdn_core.__file__)):
        try:
            out[name] = _file_sha(Path(f))
        except Exception:
            out[name] = "?"
    return out


def fold_of_path(path: Path) -> int | None:
    """k of the innermost fold_k directory of the path (the file name itself is ignored)."""
    ks = [int(m.group(1)) for part in Path(path).parts[:-1] for m in [FOLD_DIR_RE.fullmatch(part)] if m]
    return ks[-1] if ks else None


def dedupe_same_file(paths) -> list[Path]:
    """One path per distinct file. The same attached output can be visible under two mount points (legacy
    /kaggle/input/<slug>/... and /kaggle/input/notebooks/<owner>/<slug>/..., or a symlink between them): paths with the
    same real path or the same content collapse to the shortest (then lexicographically first) one, so every fold picks
    its file from the same mount."""
    ordered = sorted((Path(p) for p in paths), key=lambda q: (len(q.parts), str(q)))
    by_real: dict[str, Path] = {}
    for p in ordered:
        by_real.setdefault(os.path.realpath(p), p)
    out = list(by_real.values())
    if len(out) > 1:
        by_sha: dict[str, Path] = {}
        for p in out:
            by_sha.setdefault(_file_sha(p), p)
        out = list(by_sha.values())
    return out


def discover_checkpoints(root: Path, name: str = "best.pt", prefer: str = "") -> dict[int, Path]:
    root = Path(root)
    by_fold: dict[int, list[Path]] = {}
    for p in find_files(root, name):
        k = fold_of_path(p.relative_to(root))
        if k is not None:
            by_fold.setdefault(k, []).append(p)
    ckpts = {}
    for k, paths in sorted(by_fold.items()):
        paths = dedupe_same_file(paths)
        if len(paths) > 1 and prefer:
            paths = [p for p in paths if prefer in str(p)] or paths
        if len(paths) > 1:
            raise RuntimeError(f"fold {k}: {len(paths)} different checkpoints {[str(p) for p in paths]}; set "
                               "KDS_CKPT_ROOT to the one trainer output to use")
        ckpts[k] = paths[0]
    # the fold models must come from ONE trainer run (folds/ dir): mixing runs breaks the video -> fold routing
    runs = sorted({str(p.parent.parent) for p in ckpts.values()})
    if len(runs) > 1:
        raise RuntimeError(f"fold checkpoints come from {len(runs)} different trainer outputs {runs}; set "
                           "KDS_CKPT_ROOT to the one trainer output to use")
    return ckpts


def inspect_checkpoints(ckpts: dict[int, Path]) -> dict[int, dict]:
    """Validate every fold checkpoint (fold id, classes, identical input contract) and record provenance."""
    import torch

    info, contracts = {}, {}
    for k, p in sorted(ckpts.items()):
        payload = torch.load(Path(p), map_location="cpu", weights_only=False)
        if not isinstance(payload, dict) or "model_state" not in payload:
            raise RuntimeError(f"{p}: not a K-DN checkpoint (dict with model_state)")
        cfg = payload.get("config") or {}
        if "fold" in payload and int(payload["fold"]) != int(k):
            raise RuntimeError(f"{p}: payload fold {payload['fold']} != directory fold_{k}")
        if cfg.get("classes") is not None and list(cfg["classes"]) != list(CLASSES):
            raise RuntimeError(f"{p}: classes {cfg['classes']} != {list(CLASSES)}")
        contracts[k] = json.dumps({key: cfg.get(key) for key in ("input", "normalization", "model")}, sort_keys=True)
        train = cfg.get("train") or {}
        if train.get("fold") is not None and int(train["fold"]) != int(k):
            raise RuntimeError(f"{p}: config.train.fold {train['fold']} != directory fold_{k}")
        info[k] = {"path": str(p), "fold": int(k), "epoch": payload.get("epoch"), "val_loss": payload.get("val_loss"),
                   "sha256": _file_sha(Path(p)), "bytes": Path(p).stat().st_size,
                   "early_stop_fold": train.get("early_stop_fold"), "n_folds": train.get("n_folds"),
                   "inner_val": train.get("inner_val")}
    if len(set(contracts.values())) > 1:
        raise RuntimeError(f"fold checkpoints disagree on the input contract: {contracts}")
    return info


def _embedded_splits_path() -> Path | None:
    here = Path(__file__).resolve().parent
    for p in (here / "splits.json", here.parent / "splits.json"):
        if p.is_file():
            return p
    return None


def resolve_splits(cfg: dict, ckpt_paths: list[Path], input_root: Path, return_sources: bool = False):
    """Video -> fold: the trainer's splits_used.json wins (it is what the fold models were trained with); videos it does
    not list fall back to the embedded splits.json. return_sources=True also returns the (primary, fallback) maps."""
    primary_path = None
    if cfg.get("splits"):
        primary_path = Path(cfg["splits"])
        if not primary_path.is_file():
            raise FileNotFoundError(f"KDS_SPLITS={primary_path} does not exist")
    else:
        cands = []
        for p in ckpt_paths:  # trainer layout: <work>/folds/fold_k/best.pt and <work>/splits_used.json
            for up in list(Path(p).parents)[:4]:
                if (up / "splits_used.json").is_file():
                    cands.append(up / "splits_used.json")
                    break
        cands = dedupe_same_file(dict.fromkeys(cands)) or dedupe_same_file(find_files(input_root, "splits_used.json"))
        if len(cands) > 1:
            cands = [c for c in cands if cfg["train_slug"] in str(c)] or cands
        if len(cands) > 1:
            raise RuntimeError(f"several different splits_used.json {[str(c) for c in cands]}; set KDS_SPLITS")
        primary_path = cands[0] if cands else None
    primary = load_splits(primary_path) if primary_path else None
    emb_path = _embedded_splits_path()
    fallback = load_splits(emb_path) if emb_path else None
    folds: dict[str, int] = dict(fallback or {})
    folds.update(primary or {})
    if not folds:
        raise RuntimeError("no splits: neither a trainer splits_used.json nor the embedded splits.json")
    disagree = sorted(s for s in (primary or {}) if fallback and s in fallback and fallback[s] != primary[s])
    info = {"primary": str(primary_path) if primary_path else None, "n_primary": len(primary or {}),
            "fallback": str(emb_path) if emb_path else None, "n_fallback": len(fallback or {}),
            "fallback_only": sorted(set(fallback or {}) - set(primary or {})) if primary else [],
            "disagreements_primary_vs_fallback": disagree}
    if primary_path:
        try:
            meta = json.loads(Path(primary_path).read_text())
            info["primary_source"] = meta.get("source") if isinstance(meta, dict) else None
            info["primary_n_folds"] = meta.get("n_folds") if isinstance(meta, dict) else None
        except Exception:
            pass
    if return_sources:
        return folds, info, primary, fallback
    return folds, info


def leak_groups(path: Path | None) -> list[list[str]]:
    """splits.json meta.leak_groups: videos that share GT annotations (overlapping space-time crops of 44b6)."""
    try:
        meta = json.loads(Path(path).read_text()).get("meta") or {}
        return [[str(s) for s in g] for g in (meta.get("leak_groups") or [])]
    except Exception:
        return []


def audit_oof_routing(video_fold: dict[str, int], ckpt_info: dict[int, dict], primary: dict[str, int] | None,
                      fallback: dict[str, int] | None, groups: list[list[str]]) -> dict:
    """Honest-OOF audit, independent of the routing itself. Fold model k trained on every video of the trainer's splits
    (splits_used.json; the embedded splits.json if the trainer's file is missing) whose fold is not k. A video v routed
    to model k = video_fold[v] is EXCLUDED (no OOF prediction, NaN) when model k trained on v or on any video of v's
    leak group (videos sharing GT annotations). Also flags folds whose best.pt epoch was selected on the OOF fold's own
    labels (trainer KDN_INNER_VAL=0: early_stop_fold == fold), a mild optimism the scorer cannot remove."""
    basis = dict(primary) if primary else dict(fallback or {})
    sib: dict[str, set] = {}
    for g in groups:
        for s in g:
            sib.setdefault(s, set()).update(g)
    excluded, n_checked = {}, 0
    for v, k in video_fold.items():
        if k not in ckpt_info:
            continue
        n_checked += 1
        seen = sorted(s for s in sib.get(v, set()) | {v} if s in basis and int(basis[s]) != int(k))
        if seen:
            excluded[v] = {"fold": int(k), "model_trained_on": seen}
    return {"basis": "splits_used.json" if primary else "embedded splits.json (trainer splits_used.json not found)",
            "videos_checked": n_checked, "leak_groups": len(groups),
            "videos_in_leak_groups": int(sum(1 for v in video_fold if v in sib)),
            "excluded": excluded,
            "early_stop_on_oof_fold": sorted(int(k) for k, v in ckpt_info.items()
                                             if v.get("early_stop_fold") is not None
                                             and int(v["early_stop_fold"]) == int(k))}


def find_coords(cfg: dict, input_root: Path) -> Path:
    if cfg.get("coords"):
        p = Path(cfg["coords"])
        if not p.is_file():
            raise FileNotFoundError(f"KDS_COORDS={p} does not exist")
        return p
    cands = find_files(input_root, cfg["coords_name"])
    if len(cands) > 1:
        cands = [c for c in cands if cfg["coords_slug"] in str(c)] or cands
    if len(cands) > 1 and len({_file_sha(c) for c in cands}) > 1:
        raise RuntimeError(f"several different {cfg['coords_name']}: {[str(c) for c in cands]}; set KDS_COORDS")
    if cands:
        return cands[0]
    local = Path(__file__).resolve().parent.parent / cfg["coords_name"]  # repo: experiments/l2_daughternet/
    if local.is_file():
        return local
    raise FileNotFoundError(f"{cfg['coords_name']} not found under {input_root} (attach vitaliineibauer/biohub-l2-coords)")


def read_coords(path: Path):
    import pandas as pd

    df = pd.read_parquet(path, columns=["stem", "node_id", "t", "z", "y", "x"])
    df["stem"] = df["stem"].astype(str)
    if df.duplicated(["stem", "node_id"]).any():
        raise RuntimeError(f"{path}: (stem, node_id) is not unique")
    return df


def select_rows(coords, cfg: dict):
    """Rows to score, in input order (index = input position). Default: every row. KDS_STEMS / KDS_MAX_VIDEOS /
    KDS_ROWS_PER_VIDEO (seeded per video) are for smoke runs and tests; workers call this with the same cfg."""
    stems_all = list(dict.fromkeys(coords["stem"].tolist()))
    stems = stems_all
    if cfg.get("stems"):
        wanted = set(cfg["stems"])
        stems = [s for s in stems_all if s in wanted]
    elif cfg.get("max_videos") and cfg["max_videos"] < len(stems_all):
        srt = sorted(stems_all)
        step = len(srt) / cfg["max_videos"]
        pick = {srt[int(i * step)] for i in range(cfg["max_videos"])}
        stems = [s for s in stems_all if s in pick]
    sel = coords if len(stems) == len(stems_all) else coords[coords["stem"].isin(stems)]
    rpv = int(cfg.get("rows_per_video") or 0)
    if rpv > 0:
        keep = []
        for stem, pos in sel.groupby("stem", sort=False).indices.items():
            if len(pos) > rpv:
                rng = np.random.default_rng([int(cfg.get("seed", 0)), zlib.crc32(str(stem).encode("utf-8"))])
                pos = np.sort(rng.choice(pos, size=rpv, replace=False))
            keep.append(pos)
        sel = sel.iloc[np.sort(np.concatenate(keep))] if keep else sel.iloc[:0]
    return sel


# ----------------------------------------------------------------------------------------------------------------------
# the DivNet sample contract (tools.divnet_infer.prepare_divnet_input) split into frame preparation + crop
# ----------------------------------------------------------------------------------------------------------------------
def contract_params(config: dict | None) -> dict:
    """The exact parameter parsing of prepare_divnet_input for a bundle config."""
    cfg = config if isinstance(config, dict) else {}
    input_cfg = cfg.get("input", {}) if isinstance(cfg, dict) else {}
    norm_cfg = cfg.get("normalization", {}) if isinstance(cfg, dict) else {}
    return {
        "lags": tuple(int(v) for v in input_cfg.get("image_lags", (-1, 0, 1, 2))),
        "pool_xy": int(input_cfg.get("pool_xy", 4)),
        "shape": (int(input_cfg.get("crop_z", 16)), int(input_cfg.get("crop_yx", 32)), int(input_cfg.get("crop_yx", 32))),
        "sigma": tuple(float(v) for v in input_cfg.get("marker_sigma", (1.5, 2.0, 2.0))),
        "clip": tuple(float(v) for v in norm_cfg.get("clip", (-0.5, 6.0))),
        "lo": float(norm_cfg.get("lo_pct", 50.0)),
        "hi": float(norm_cfg.get("hi_pct", 99.5)),
    }


def prepare_frame(reader, stem: str, frame_t: int, params: dict) -> np.ndarray:
    """prepared_frames[(stem, frame_t)] of the contract: raw frame (FrameReader clamps t to [0, T-1]) -> pool -> norm."""
    raw = reader(stem, int(frame_t), {})
    return _normalise(_pool_xy(raw, params["pool_xy"]), params["lo"], params["hi"], params["clip"])


def frame_keys(t_values, lags) -> list[int]:
    return sorted({max(0, int(tt) + int(lag)) for tt in np.unique(np.asarray(t_values)) for lag in lags})


class SyncFrames:
    """Frames prepared inline (no threads)."""

    def __init__(self, reader, params: dict) -> None:
        self.reader, self.params = reader, params

    def request(self, stem: str, keys) -> None:
        pass

    def get(self, stem: str, frame_t: int) -> np.ndarray:
        return prepare_frame(self.reader, stem, frame_t, self.params)

    def drop(self, stem: str) -> None:
        pass

    def close(self) -> None:
        pass


class FramePrefetcher(SyncFrames):
    """Thread pool preparing frames in request order (FIFO), so the next video's frames are ready when it starts."""

    def __init__(self, reader, params: dict, threads: int = 2) -> None:
        super().__init__(reader, params)
        self.pool = ThreadPoolExecutor(max(1, int(threads)), thread_name_prefix="kds-frame")
        self.futures: dict[tuple, object] = {}
        self.lock = threading.Lock()
        self.busy_s = 0.0

    def _task(self, stem: str, frame_t: int) -> np.ndarray:
        t0 = time.time()
        out = prepare_frame(self.reader, stem, frame_t, self.params)
        with self.lock:
            self.busy_s += time.time() - t0
        return out

    def request(self, stem: str, keys) -> None:
        for ft in keys:
            key = (stem, int(ft))
            if key not in self.futures:
                self.futures[key] = self.pool.submit(self._task, stem, int(ft))

    def get(self, stem: str, frame_t: int) -> np.ndarray:
        fut = self.futures.pop((stem, int(frame_t)), None)
        return fut.result() if fut is not None else prepare_frame(self.reader, stem, frame_t, self.params)

    def drop(self, stem: str) -> None:
        for key in [k for k in self.futures if k[0] == stem]:
            self.futures.pop(key).cancel()

    def close(self) -> None:
        for fut in self.futures.values():
            fut.cancel()
        self.futures.clear()
        self.pool.shutdown(wait=True)


def score_video(stem: str, t, z, y, x, bundle: dict, reader, model_indices=None, **kw) -> tuple[np.ndarray, dict]:
    """Softmax (N, 3) in CLASSES order for the N nodes of one video (arrays in any order; the result is aligned to them).

    model_indices selects bundle['models'] (OOF: the one fold model of the video); None = mean over all (ensemble), the
    same reduction as tools.daughternet.score_nodes. Samples equal prepare_sample's exactly (checked per video).
    Keyword options: see score_video_members."""
    _, mean, stats = score_video_members(stem, t, z, y, x, bundle, reader, model_indices, **kw)
    return mean, stats


def score_video_members(stem: str, t, z, y, x, bundle: dict, reader, model_indices=None, *, batch: int = 512,
                        amp: bool = False, frames: SyncFrames | None = None, checks_fresh: int = 1,
                        checks_cached: int = 2, seed: int = 0,
                        pad_to: int = 64) -> tuple[np.ndarray, np.ndarray, dict]:
    """(members (M, N, 3), mean (N, 3), stats): the softmax of each selected model and their mean (score_nodes'
    torch.stack(...).mean(0), computed on the device), from ONE pass over the frames. With a single selected model the
    mean equals that model's softmax exactly; with all models, members[k] is fold model k's OOF-able output."""
    import torch

    t = np.asarray(t, dtype=np.int64)
    z, y, x = (np.asarray(v, dtype=np.float64) for v in (z, y, x))
    n = len(t)
    stats = {"rows": int(n), "frames": 0, "frame_wait_s": 0.0, "build_s": 0.0, "model_s": 0.0, "gpu_wait_s": 0.0,
             "checks_fresh": 0, "checks_cached": 0, "batches": 0}
    n_models = len(bundle["models"]) if model_indices is None else len(model_indices)
    out = np.full((n, len(CLASSES)), np.nan, dtype=np.float32)
    members = np.full((n_models, n, len(CLASSES)), np.nan, dtype=np.float32)
    if n == 0:
        return members, out, stats
    t_start = time.time()
    params = contract_params(bundle.get("config"))
    lags, shape, pool = params["lags"], params["shape"], params["pool_xy"]
    min_lag = min(lags)
    marker = _marker(shape, params["sigma"])
    models = list(bundle["models"]) if model_indices is None else [bundle["models"][int(i)] for i in model_indices]
    if not models:
        raise RuntimeError(f"{stem}: no model selected")
    dev = bundle["device"]
    use_amp = bool(amp) and dev.type == "cuda"
    frames = frames if frames is not None else SyncFrames(reader, params)
    frames.request(stem, frame_keys(t, lags))
    order = np.argsort(t, kind="stable")
    rng = np.random.default_rng([int(seed), zlib.crc32(stem.encode("utf-8"))])
    n_fresh, n_cached = min(max(0, checks_fresh), n), max(0, checks_cached)
    picks = rng.choice(n, size=min(n, n_fresh + n_cached), replace=False).tolist()
    fresh, cached = set(picks[:n_fresh]), set(picks[n_fresh:])
    prepared: dict[tuple, np.ndarray] = {}  # (stem, frame_t) -> prepared frame, the contract's prepared_frames layout
    cur_t = None
    pending = None

    def collect(item) -> None:
        idx, pm, p = item
        tw = time.time()
        arr = p.float().cpu().numpy()
        arr_m = pm.float().cpu().numpy()
        stats["gpu_wait_s"] += time.time() - tw
        stats["model_s"] += time.time() - tw
        out[idx] = arr[:len(idx)]
        members[:, idx] = arr_m[:, :len(idx)]

    for s in range(0, n, batch):
        tb = time.time()
        idx = order[s:s + batch]
        nb = len(idx)
        nb_pad = min(max(batch, nb), -(-nb // pad_to) * pad_to) if pad_to else nb
        xb = np.zeros((nb_pad, len(lags) + 1, *shape), dtype=np.float32)
        xb[:nb, len(lags)] = marker
        wait0 = stats["frame_wait_s"]
        for j, i in enumerate(idx):
            tt = int(t[i])
            if tt != cur_t:  # nodes are visited by t: frames below tt + min_lag are never needed again
                for key in [k for k in prepared if k[1] < tt + min_lag]:
                    del prepared[key]
                cur_t = tt
            centre = (int(round(float(z[i]))), int(round(float(y[i]) / pool)), int(round(float(x[i]) / pool)))
            for c, lag in enumerate(lags):
                key = (stem, max(0, tt + lag))
                frame = prepared.get(key)
                if frame is None:
                    tw = time.time()
                    frame = frames.get(stem, key[1])
                    stats["frame_wait_s"] += time.time() - tw
                    stats["frames"] += 1
                    prepared[key] = frame
                xb[j, c] = _crop_with_padding(frame, centre, shape)
            pos = s + j
            if pos in fresh or pos in cached:
                node = {"t": tt, "z": float(z[i]), "y": float(y[i]), "x": float(x[i])}
                ref_bundle = {"config": bundle.get("config"), "prepared_frames": {} if pos in fresh else prepared}
                ref = prepare_sample(stem, tt, node, ref_bundle, {}, reader)
                if ref.shape != xb[j].shape or not np.array_equal(ref, xb[j]):
                    diff = float(np.abs(ref - xb[j]).max()) if ref.shape == xb[j].shape else float("nan")
                    raise ContractError(f"{stem} node t={tt} z={z[i]} y={y[i]} x={x[i]}: fast-path sample != "
                                        f"prepare_sample ({'fresh' if pos in fresh else 'cached'} frames, max |d| {diff})")
                stats["checks_fresh" if pos in fresh else "checks_cached"] += 1
        if not np.isfinite(xb[:nb]).all():
            raise RuntimeError(f"{stem}: non-finite DaughterNet input in batch starting at sorted row {s}")
        stats["build_s"] += time.time() - tb - (stats["frame_wait_s"] - wait0)
        tf = time.time()  # on CUDA this is H2D + kernel launch (async); on CPU it is the whole forward pass
        xt = torch.from_numpy(xb).to(dev)
        with torch.no_grad(), torch.autocast(device_type=dev.type, dtype=torch.float16, enabled=use_amp):
            pm = torch.stack([torch.softmax(m(xt).float(), dim=1) for m in models])
            p = pm.mean(0)
        stats["model_s"] += time.time() - tf
        if pending is not None:
            collect(pending)
        pending = (idx, pm, p)
        stats["batches"] += 1
    collect(pending)
    prepared.clear()
    stats["secs"] = time.time() - t_start
    stats["models"] = len(models)
    return members, out, stats


# ----------------------------------------------------------------------------------------------------------------------
# parts (per-video results, resumable)
# ----------------------------------------------------------------------------------------------------------------------
def part_path(parts_dir: Path, stem: str) -> Path:
    return Path(parts_dir) / f"{stem}.parquet"


def write_part(df, path: Path) -> None:
    tmp = path.with_name(path.name + ".tmp")
    df.to_parquet(tmp, index=False)
    os.replace(tmp, path)


def part_valid(path: Path, node_ids: np.ndarray) -> bool:
    import pandas as pd

    if not path.is_file():
        return False
    try:
        df = pd.read_parquet(path)
    except Exception:
        return False
    if list(df.columns) != OUT_COLUMNS or len(df) != len(node_ids):
        return False
    if not np.array_equal(df["node_id"].to_numpy(np.int64), np.asarray(node_ids, dtype=np.int64)):
        return False
    return bool(np.isfinite(df[P_COLUMNS].to_numpy(np.float64)).all())


# ----------------------------------------------------------------------------------------------------------------------
# worker process (one per GPU)
# ----------------------------------------------------------------------------------------------------------------------
def worker_main(argv: list[str]) -> int:
    """Score the videos assigned to this worker for the modes the job lists per video. A video that needs 'ens' runs all
    fold models once; its 'oof' part (if also requested) is fold model k's softmax from the same pass."""
    import argparse

    import pandas as pd
    import torch

    ap = argparse.ArgumentParser()
    ap.add_argument("--job", required=True)
    ap.add_argument("--worker", type=int, required=True)
    ap.add_argument("--device", default="cpu")
    ap.add_argument("--threads", type=int, default=0)
    a = ap.parse_args(argv)
    job = json.loads(Path(a.job).read_text())
    cfg = job["cfg"]
    stems = job["workers"][str(a.worker)]
    stem_modes = job["stem_modes"]
    parts_dirs = {m: Path(d) for m, d in job["parts_dirs"].items()}
    if a.threads:
        torch.set_num_threads(int(a.threads))
    try:
        import blosc2
        blosc2.set_nthreads(max(1, int(job.get("blosc_threads", 1))))
    except Exception:
        pass
    device = torch.device(a.device)
    if device.type == "cuda":
        torch.backends.cudnn.benchmark = True
    t0 = time.time()
    sel = select_rows(read_coords(Path(job["coords"])), cfg)
    sel = sel[sel["stem"].isin(set(stems))]
    groups = {s: g for s, g in sel.groupby("stem", sort=False)}
    ckpts = {int(k): v for k, v in job["ckpts"].items()}
    model_folds = sorted(ckpts)
    bundle = load_daughternet([ckpts[f] for f in model_folds], device=str(device))
    fold_index = {f: i for i, f in enumerate(model_folds)}
    reader = FrameReader(Path(job["data_dir"]))
    params = contract_params(bundle["config"])
    frames = (FramePrefetcher(reader, params, cfg["frame_threads"]) if int(cfg["frame_threads"]) > 0
              else SyncFrames(reader, params))
    run_dir = Path(job["run_dir"])
    name = torch.cuda.get_device_name(0) if device.type == "cuda" else "cpu"
    log(f"[w{a.worker}] {len(stems)} videos, {len(sel)} rows, modes {sorted({m for s in stems for m in stem_modes[s]})}, "
        f"models {model_folds} on {device} ({name}), batch {cfg['batch']}, amp {cfg['amp']}, frame threads "
        f"{cfg['frame_threads']}; setup {time.time() - t0:.1f} s")
    records, rc = [], 0
    try:
        for vi, stem in enumerate(stems):
            g = groups.get(stem)
            modes_s = list(stem_modes[stem])
            if g is None or g.empty:
                records.append({"stem": stem, "rows": 0, "modes": modes_s, "error": "no rows"})
                continue
            frames.request(stem, frame_keys(g["t"].to_numpy(), params["lags"]))
            if vi + 1 < len(stems) and stems[vi + 1] in groups:
                frames.request(stems[vi + 1], frame_keys(groups[stems[vi + 1]]["t"].to_numpy(), params["lags"]))
            fold = int(job["folds"][stem])
            joint = "ens" in modes_s
            model_idx = None if joint else [fold_index[fold]]
            try:
                members, mean, st = score_video_members(
                    stem, g["t"].to_numpy(), g["z"].to_numpy(), g["y"].to_numpy(), g["x"].to_numpy(), bundle, reader,
                    model_idx, batch=int(cfg["batch"]), amp=bool(cfg["amp"]), frames=frames,
                    checks_fresh=int(cfg["checks_fresh"]), checks_cached=int(cfg["checks_cached"]),
                    seed=int(cfg["seed"]), pad_to=int(cfg["pad_to"]))
            except ContractError as exc:
                log(f"[w{a.worker}] CONTRACT FAILURE {exc}")
                records.append({"stem": stem, "rows": int(len(g)), "modes": modes_s, "error": f"ContractError: {exc}"})
                rc = 3
                break
            except Exception as exc:
                frames.drop(stem)
                log(f"[w{a.worker}] {stem} FAILED {type(exc).__name__}: {exc}")
                records.append({"stem": stem, "rows": int(len(g)), "modes": modes_s,
                                "error": f"{type(exc).__name__}: {exc}"})
                rc = rc or 1
                continue
            for mode in modes_s:
                p = mean if (mode == "ens" or not joint) else members[fold_index[fold]]
                part = pd.DataFrame({"stem": stem, "node_id": g["node_id"].to_numpy(np.int64),
                                     "fold": np.full(len(g), fold, dtype=np.int16),
                                     "p_cont": p[:, 0], "p_par": p[:, 1], "p_dau": p[:, 2]})[OUT_COLUMNS]
                write_part(part, part_path(parts_dirs[mode], stem))
            frames.drop(stem)
            records.append({"stem": stem, "fold": fold, "modes": modes_s,
                            **{k: (round(v, 3) if isinstance(v, float) else v) for k, v in st.items()}, "error": None})
            log(f"[w{a.worker}] {vi + 1}/{len(stems)} {stem} fold {fold} {'+'.join(modes_s)} rows {st['rows']} x "
                f"{st['models']} models, frames {st['frames']} {st['secs']:.1f}s (build {st['build_s']:.1f} frame-wait "
                f"{st['frame_wait_s']:.1f} model {st['model_s']:.1f} of which gpu-wait {st['gpu_wait_s']:.1f}) "
                f"{st['rows'] / max(st['secs'], 1e-9):.0f} rows/s checks {st['checks_fresh']}+{st['checks_cached']}")
    finally:
        busy = getattr(frames, "busy_s", None)
        frames.close()
        secs = time.time() - t0
        rows_done = sum(r.get("rows", 0) for r in records if not r.get("error"))
        summary = {"worker": a.worker, "device": str(device), "device_name": name, "attempt": job.get("attempt", 0),
                   "videos": len(stems), "rows": int(rows_done), "secs": round(secs, 1),
                   "rows_per_s": round(rows_done / max(secs, 1e-9), 1), "frame_thread_busy_s": busy,
                   "decoders": decoder_names(), "records": records}
        (run_dir / f"_worker_{a.worker}_attempt{job.get('attempt', 0)}.json").write_text(
            json.dumps(summary, indent=1, default=_json_default))
    log(f"[w{a.worker}] done: {summary['rows']} rows in {secs:.0f} s ({summary['rows_per_s']} rows/s) rc={rc}")
    return rc


# ----------------------------------------------------------------------------------------------------------------------
# driver
# ----------------------------------------------------------------------------------------------------------------------
def _json_default(value):
    if isinstance(value, np.integer):
        return int(value)
    if isinstance(value, np.floating):
        return float(value)
    if isinstance(value, np.ndarray):
        return value.tolist()
    return str(value)


def _slots(cfg: dict) -> list[str]:
    import torch

    n_gpu = torch.cuda.device_count() if cfg["gpus"] != 0 else 0
    if cfg["gpus"] > 0:
        n_gpu = min(n_gpu, cfg["gpus"])
    if n_gpu:
        n = cfg["workers"] or min(n_gpu, max(1, cfg["max_gpu_workers"]))
        return [str(i % n_gpu) for i in range(n)]
    return ["cpu"] * max(1, cfg["workers"] or 1)


def plan_workers(sel, stems: list[str], n_workers: int, lags, n_models: dict[str, int] | None = None
                 ) -> dict[int, list[str]]:
    """Greedy longest-first on rows x models + FRAME_COST_ROWS x prepared frames."""
    t_by_stem = {s: g for s, g in sel.groupby("stem", sort=False)["t"]}
    n_models = n_models or {}
    cost = {s: len(t_by_stem[s]) * n_models.get(s, 1) + FRAME_COST_ROWS * len(frame_keys(t_by_stem[s].to_numpy(), lags))
            for s in stems}
    load = [0.0] * n_workers
    out: dict[int, list[str]] = {w: [] for w in range(n_workers)}
    for s in sorted(stems, key=lambda k: (-cost[k], k)):
        w = min(range(n_workers), key=lambda k: (load[k], k))
        out[w].append(s)
        load[w] += cost[s]
    return out


def _launch(job_path: Path, assignment: dict[int, list[str]], slots: list[str], log_dir: Path) -> dict[int, int]:
    env_base = os.environ.copy()
    env_base["PYTHONPATH"] = os.pathsep.join(SRC_PATHS + [p for p in env_base.get("PYTHONPATH", "").split(os.pathsep) if p])
    env_base["PYTHONUNBUFFERED"] = "1"
    threads = max(1, (os.cpu_count() or 1) // max(1, len(slots)))
    running = {}
    for wid, slot in enumerate(slots):
        if not assignment.get(wid):
            continue
        env = dict(env_base)
        env["CUDA_VISIBLE_DEVICES"] = slot if slot != "cpu" else ""
        device = "cuda:0" if slot != "cpu" else "cpu"
        cmd = [sys.executable, str(Path(__file__).resolve()), "worker", "--job", str(job_path), "--worker", str(wid),
               "--device", device, "--threads", str(threads)]
        sink = open(log_dir / f"worker_{wid}.log", "a", encoding="utf-8")
        proc = subprocess.Popen(cmd, env=env, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True,
                                encoding="utf-8", errors="replace")
        th = threading.Thread(target=kdn_core._pump, args=(proc.stdout, f"[w{wid}@{slot}] ", sink), daemon=True)
        th.start()
        running[wid] = (proc, th, sink)
        log(f"[score] worker {wid} on {slot}: {len(assignment[wid])} videos (pid {proc.pid})")
    rcs = {}
    for wid, (proc, th, sink) in running.items():
        rcs[wid] = proc.wait()
        th.join(timeout=10)
        sink.close()
    return rcs


def _check_signature(parts_dir: Path, signature: dict, resume: bool) -> str:
    sig_path = parts_dir / "_signature.json"
    old = json.loads(sig_path.read_text()) if sig_path.is_file() else None
    stale = [p for p in parts_dir.glob("*.parquet")]
    status = "fresh"
    if stale and (not resume or old != signature):
        for p in stale:
            p.unlink()
        status = "cleared (resume off)" if not resume else "cleared (signature changed)"
    elif stale:
        status = "resume"
    sig_path.write_text(json.dumps(signature, indent=1))
    return status


def verify_rows(out, sel, ckpts: dict[int, Path], mode: str, data_dir: Path, per_fold: int, seed: int) -> dict:
    """Re-score a few rows per fold from raw frames through tools.daughternet.score_nodes (CPU, batch 1)."""
    reader = FrameReader(Path(data_dir))
    rng = np.random.default_rng(seed)
    both = out.join(sel[["t", "z", "y", "x"]])
    both = both[np.isfinite(both[P_COLUMNS].to_numpy(np.float64)).all(axis=1)]
    picks = []
    for fold, g in both.groupby("fold"):
        cand = [g["p_par"].idxmax(), g["p_dau"].idxmax()] + rng.choice(g.index.to_numpy(), size=min(len(g), per_fold),
                                                                       replace=False).tolist()
        picks += list(dict.fromkeys(cand))[:max(0, per_fold)]
    bundles, rows, max_diff = {}, [], 0.0
    for i in picks:
        r = both.loc[i]
        folds = (int(r["fold"]),) if mode == "oof" else tuple(sorted(ckpts))
        if folds not in bundles:
            bundles[folds] = load_daughternet([ckpts[f] for f in folds], device="cpu")
        bundle = bundles[folds]
        bundle["prepared_frames"] = {}
        node = {"t": int(r["t"]), "z": float(r["z"]), "y": float(r["y"]), "x": float(r["x"])}
        p = score_nodes(str(r["stem"]), [node], bundle, {}, reader, batch=1)[0]
        diff = float(np.abs(p.astype(np.float64) - r[P_COLUMNS].to_numpy(np.float64)).max())
        max_diff = max(max_diff, diff)
        rows.append({"row": int(i), "stem": str(r["stem"]), "node_id": int(r["node_id"]), "fold": int(r["fold"]),
                     "models": list(folds), "p_par": float(r["p_par"]), "diff": diff})
    return {"checked": len(rows), "max_abs_diff": max_diff, "rows": rows}


def _worker_stats(run_dir: Path) -> list[dict]:
    workers = []
    for p in sorted(run_dir.glob("_worker_*.json")):
        try:
            w = json.loads(p.read_text())
        except Exception:
            continue
        recs = [r for r in w.get("records", []) if not r.get("error")]
        workers.append({k: w.get(k) for k in ("worker", "attempt", "device", "device_name", "videos", "rows", "secs",
                                              "rows_per_s", "frame_thread_busy_s", "decoders")}
                       | {f"sum_{k}": round(sum(float(r.get(k, 0) or 0) for r in recs), 1)
                          for k in ("frame_wait_s", "build_s", "model_s", "gpu_wait_s", "frames", "checks_fresh",
                                    "checks_cached")}
                       | {"failed": [r for r in w.get("records", []) if r.get("error")]})
    return workers


def score_modes(cfg: dict, modes: list[str], ctx: dict) -> dict:
    """One worker pass for all requested modes (a video needing oof + ens runs every fold model once and yields both).
    Returns per-mode bookkeeping for finish_mode."""
    work, sel, ckpts = ctx["work"], ctx["sel"], ctx["ckpts"]
    run_dir = work / "parts" / "_run"
    run_dir.mkdir(parents=True, exist_ok=True)
    for old in list(run_dir.glob("_worker_*.json")) + list(run_dir.glob("_job_*.json")):
        old.unlink()  # per-run bookkeeping; the summaries describe this run's workers only
    node_ids, video_fold = ctx["node_ids"], ctx["video_fold"]
    book = {}
    for mode in modes:
        parts_dir = work / "parts" / mode
        parts_dir.mkdir(parents=True, exist_ok=True)
        signature = {"mode": mode, "classes": list(CLASSES), "coords_sha256": ctx["coords_sha"], "amp": bool(cfg["amp"]),
                     "ckpt_sha256": {str(k): v["sha256"] for k, v in ctx["ckpt_info"].items()},
                     "code_sha256": ctx.get("code_sha", {}),
                     "folds": {s: video_fold[s] for s in ctx["stems"]}}
        status = _check_signature(parts_dir, signature, cfg["resume"])
        no_model = [s for s in ctx["stems"] if (video_fold[s] not in ckpts if mode == "oof" else not ckpts)]
        # honest OOF: a video whose fold model trained on it or on a leak-group sibling gets no OOF prediction
        leak_excluded = [s for s in ctx["stems"] if mode == "oof" and s in ctx["routing"]["excluded"]
                         and s not in set(no_model)]
        skip = set(no_model) | set(leak_excluded)
        candidates = [s for s in ctx["stems"] if s not in skip]
        if leak_excluded:  # a part from an earlier run must never be merged for an excluded video
            for s in leak_excluded:
                part_path(parts_dir, s).unlink(missing_ok=True)
        todo = [s for s in candidates if not part_valid(part_path(parts_dir, s), node_ids[s])]
        book[mode] = {"parts_dir": parts_dir, "status": status, "no_model": no_model, "leak_excluded": leak_excluded,
                      "candidates": candidates, "todo_start": list(todo), "todo": list(todo),
                      "resumed": len(candidates) - len(todo)}
        log(f"[score:{mode}] parts {status}: {book[mode]['resumed']} videos resumed, {len(todo)} to score, "
            f"{len(no_model)} without a model {no_model[:5]}, {len(leak_excluded)} excluded by the leak audit "
            f"{leak_excluded[:5]}")
    slots = ctx["slots"]
    rcs_all = []
    t0 = time.time()
    for attempt in range(1 + max(0, int(cfg["retries"]))):
        stem_modes: dict[str, list[str]] = {}
        for mode in modes:
            for s in book[mode]["todo"]:
                stem_modes.setdefault(s, []).append(mode)
        if not stem_modes:
            break
        todo = [s for s in ctx["stems"] if s in stem_modes]
        assignment = plan_workers(sel, todo, len(slots), ctx["lags"],
                                  {s: (len(ckpts) if "ens" in m else 1) for s, m in stem_modes.items()})
        job = {"cfg": dict(cfg), "attempt": attempt, "coords": str(ctx["coords"]), "data_dir": str(ctx["data_dir"]),
               "run_dir": str(run_dir), "parts_dirs": {m: str(book[m]["parts_dir"]) for m in modes},
               "ckpts": {str(k): str(p) for k, p in ckpts.items()}, "folds": {s: video_fold[s] for s in todo},
               "stem_modes": stem_modes, "workers": {str(w): v for w, v in assignment.items()},
               "blosc_threads": max(1, (os.cpu_count() or 1) // max(1, len(slots) * max(1, cfg["frame_threads"])))}
        job_path = run_dir / f"_job_attempt{attempt}.json"
        job_path.write_text(json.dumps(job, indent=1))
        combos: dict[str, int] = {}
        for m in stem_modes.values():
            combos["+".join(m)] = combos.get("+".join(m), 0) + 1
        log(f"[score] attempt {attempt}: {len(todo)} videos (modes per video {combos}), {len(slots)} workers")
        rcs = _launch(job_path, assignment, slots, run_dir)
        rcs_all.append({str(k): v for k, v in rcs.items()})
        if any(rc == 3 for rc in rcs.values()):
            raise ContractError(f"a worker hit a contract failure (see its log in {run_dir}); no merged output written")
        for mode in modes:
            book[mode]["todo"] = [s for s in book[mode]["todo"]
                                  if not part_valid(part_path(book[mode]["parts_dir"], s), node_ids[s])]
        log(f"[score] attempt {attempt}: worker rc {rcs}; videos still missing "
            f"{ {m: len(book[m]['todo']) for m in modes} }")
    score_s = time.time() - t0
    workers = _worker_stats(run_dir)
    for mode in modes:
        book[mode].update({"score_s": score_s, "worker_rc": rcs_all, "workers": workers, "joint_modes": list(modes)})
    return book


def finish_mode(cfg: dict, mode: str, ctx: dict, bk: dict) -> dict:
    """Merge the parts of one mode into node_p_<mode>.parquet (input order), verify, write the summary json."""
    import pandas as pd

    t0 = time.time()
    work, sel, ckpts = ctx["work"], ctx["sel"], ctx["ckpts"]
    node_ids, video_fold = ctx["node_ids"], ctx["video_fold"]
    parts_dir, todo = bk["parts_dir"], set(bk["todo"])
    parts = [pd.read_parquet(part_path(parts_dir, s)) for s in bk["candidates"] if s not in todo]
    res = pd.concat(parts, ignore_index=True) if parts else pd.DataFrame(columns=OUT_COLUMNS)
    res = res.drop(columns=["fold"])
    out = (sel[["stem", "node_id"]].rename_axis("_row").reset_index()
           .merge(res, on=["stem", "node_id"], how="left", validate="one_to_one").set_index("_row"))
    out.index.name = None
    out["fold"] = out["stem"].map(video_fold).astype(np.int16)
    out = out[OUT_COLUMNS]
    for c in P_COLUMNS:
        out[c] = out[c].astype(np.float32)
    if len(out) != len(sel) or not (out.index == sel.index).all() or not (out["node_id"].to_numpy() ==
                                                                         sel["node_id"].to_numpy()).all():
        raise RuntimeError("merged output is not aligned to the input rows")
    out_path = work / OUTPUTS[mode]
    tmp = out_path.with_name(out_path.name + ".tmp")
    out.reset_index(drop=True).to_parquet(tmp, index=False)
    os.replace(tmp, out_path)
    merge_s = time.time() - t0

    tv = time.time()
    tol = cfg["verify_tol"] or (2e-2 if cfg["amp"] else 1e-4)
    try:
        verify = verify_rows(out, sel, ckpts, mode, ctx["data_dir"], cfg["verify_per_fold"], cfg["seed"])
        verify["tolerance"] = tol
        verify["pass"] = bool(verify["checked"] > 0 and verify["max_abs_diff"] <= tol)
    except Exception as exc:
        verify = {"checked": 0, "error": f"{type(exc).__name__}: {exc}", "pass": False, "tolerance": tol}
    verify_s = time.time() - tv

    scored_now = [s for s in bk["todo_start"] if s not in todo]
    rows_now = int(sum(len(node_ids[s]) for s in scored_now))
    p = out[P_COLUMNS].to_numpy(np.float64)
    finite = np.isfinite(p).all(axis=1)
    nan_rows = int((~finite).sum())
    score_s = bk["score_s"]
    summary = {
        "kernel": KERNEL, "mode": mode, "output": str(out_path), "columns": OUT_COLUMNS, "classes": list(CLASSES),
        "rows_input_parquet": int(ctx["n_coords"]), "rows_selected": int(len(sel)), "rows_out": int(len(out)),
        "row_count_matches_input": bool(len(out) == ctx["n_coords"]), "selection": ctx["selection"],
        "nan_rows": nan_rows,
        "nan_rows_per_fold": {str(k): int(v) for k, v in out.loc[~finite, "fold"].value_counts().sort_index().items()},
        "rows_per_fold": {str(k): int(v) for k, v in out["fold"].value_counts().sort_index().items()},
        "videos_per_fold": {str(k): int(v) for k, v in
                            pd.Series([video_fold[s] for s in ctx["stems"]]).value_counts().sort_index().items()},
        "models_per_row": 1 if mode == "oof" else len(ckpts),
        "fold_column": ("split fold of the video = the fold model that scored the row (never trained on the video)"
                        if mode == "oof" else "split fold of the video, for reference; p averages all fold models"),
        "videos": {"total": len(ctx["stems"]), "resumed": bk["resumed"], "scored_this_run": len(scored_now),
                   "rows_scored_this_run": rows_now, "missing_after_retries": sorted(todo), "no_model": bk["no_model"],
                   "leak_excluded": bk.get("leak_excluded", [])},
        "worker_rc": bk["worker_rc"], "parts_status": bk["status"],
        "checkpoints": {str(k): v for k, v in ctx["ckpt_info"].items()}, "splits": ctx["splits_info"],
        "oof_routing_audit": ctx["routing"] if mode == "oof" else None,
        "prob": {"max_abs_sum_minus_1": float(np.abs(p[finite].sum(axis=1) - 1).max()) if finite.any() else None,
                 "mean": {c: float(np.nanmean(p[:, i])) if finite.any() else None for i, c in enumerate(CLASSES)},
                 "p_par_quantiles": {q: float(np.quantile(p[finite, 1], float(q))) for q in ("0.5", "0.9", "0.99", "0.999")}
                 if finite.any() else None},
        "verify": verify,
        "timing": {"discover_s": ctx["discover_s"], "score_wall_s": round(score_s, 1), "merge_s": round(merge_s, 1),
                   "verify_s": round(verify_s, 1), "finish_s": round(time.time() - t0, 1),
                   "rows_per_s_wall": round(rows_now / max(score_s, 1e-9), 1) if rows_now else None,
                   "joint_pass_modes": bk["joint_modes"], "workers": bk["workers"]},
        "slots": ctx["slots"], "config": cfg,
    }
    sum_path = work / f"node_p_{mode}_summary.json"
    sum_path.write_text(json.dumps(summary, indent=1, default=_json_default))
    log(f"[score:{mode}] wrote {out_path} ({len(out)} rows, NaN rows {nan_rows}) and {sum_path}; rows per fold "
        f"{summary['rows_per_fold']}; merge {merge_s:.1f} s; verify {verify.get('checked')} rows "
        f"max |dp| {verify.get('max_abs_diff')} tol {tol} -> {'PASS' if verify['pass'] else 'FAIL'}")
    return summary


def run(cfg: dict | None = None) -> dict:
    cfg = cfg or config_from_env()
    t0 = time.time()
    work = Path(cfg["work_dir"])
    work.mkdir(parents=True, exist_ok=True)
    input_root = Path(cfg["input_root"])
    data_dir = resolve_data_dir({"data_dir": cfg["data_dir"]})
    ckpt_root = Path(cfg["ckpt_root"]) if cfg.get("ckpt_root") else input_root
    ckpts = discover_checkpoints(ckpt_root, cfg["ckpt_name"], cfg["train_slug"])
    if not ckpts:
        raise FileNotFoundError(f"no fold_k/{cfg['ckpt_name']} under {ckpt_root} (attach the K-DN trainer output)")
    ckpt_info = inspect_checkpoints(ckpts)
    log("[discover] checkpoints: " + "; ".join(f"fold {k}: {v['path']} epoch {v['epoch']} val {v['val_loss']}"
                                              for k, v in ckpt_info.items()))
    if cfg["expect_folds"] and len(ckpts) != cfg["expect_folds"]:
        log(f"[discover] WARNING {len(ckpts)} fold checkpoints found, expected {cfg['expect_folds']}")
    folds, splits_info, primary, fallback = resolve_splits(cfg, list(ckpts.values()), input_root, return_sources=True)
    basis = primary or fallback or {}
    basis_nf = int(splits_info.get("primary_n_folds") or 0) if primary else 0
    basis_nf = basis_nf or len(set(basis.values()))
    ck_nf = sorted({int(v["n_folds"]) for v in ckpt_info.values() if v.get("n_folds") is not None})
    if ck_nf and ck_nf != [basis_nf]:
        raise RuntimeError(f"fold checkpoints were trained with n_folds {ck_nf} but the splits used for routing "
                           f"({splits_info['primary'] or splits_info['fallback']}) have {basis_nf} folds: the "
                           "video -> fold model routing would not be out-of-fold; set KDS_SPLITS to the trainer's "
                           "splits_used.json")
    if not primary:
        log("[discover] WARNING no trainer splits_used.json found: routing assumes the trainer used the embedded "
            "splits.json (fold counts agree)")
    elif splits_info["fallback_only"]:
        log(f"[discover] WARNING {len(splits_info['fallback_only'])} videos are not in the trainer's splits_used.json "
            "and are routed by the embedded splits.json (no fold model trained on them)")
    coords_path = find_coords(cfg, input_root)
    coords = read_coords(coords_path)
    sel = select_rows(coords, cfg)
    stems = list(dict.fromkeys(sel["stem"].tolist()))
    no_data = [s for s in stems if not (data_dir / f"{s}.zarr").exists()]
    if no_data:
        log(f"[discover] WARNING {len(no_data)} videos have no {data_dir}/<stem>.zarr: {no_data[:5]}")
    import torch
    bundle_cfg = torch.load(ckpts[min(ckpts)], map_location="cpu", weights_only=False).get("config")
    lags = contract_params(bundle_cfg)["lags"]
    slots = _slots(cfg)
    selection = {"stems": cfg["stems"] or None, "max_videos": cfg["max_videos"], "rows_per_video": cfg["rows_per_video"],
                 "seed": cfg["seed"], "videos": len(stems)}
    log(f"[discover] coords {coords_path}: {len(coords)} rows / {coords['stem'].nunique()} videos; selected {len(sel)} "
        f"rows / {len(stems)} videos; data {data_dir}; splits primary {splits_info['primary']} "
        f"(fallback-only {len(splits_info['fallback_only'])}, disagreements "
        f"{len(splits_info['disagreements_primary_vs_fallback'])}); worker slots {slots}")
    video_fold = {s: int(folds.get(s, -1)) for s in stems}
    routing = audit_oof_routing(video_fold, ckpt_info, primary, fallback, leak_groups(_embedded_splits_path()))
    if routing["excluded"]:
        log(f"[discover] WARNING OOF leak audit: {len(routing['excluded'])} videos would be scored by a fold model that "
            f"trained on them or on a leak-group sibling -> NaN in node_p_oof: {dict(list(routing['excluded'].items())[:3])}")
    if routing["early_stop_on_oof_fold"]:
        log(f"[discover] note: folds {routing['early_stop_on_oof_fold']} selected best.pt on their own OOF fold "
            "(trainer KDN_INNER_VAL=0), so OOF p is mildly optimistic")
    ctx = {"work": work, "sel": sel, "folds": folds, "ckpts": ckpts, "ckpt_info": ckpt_info, "stems": stems,
           "coords": coords_path, "coords_sha": _file_sha(coords_path), "n_coords": len(coords), "data_dir": data_dir,
           "splits_info": splits_info, "slots": slots, "lags": lags, "selection": selection,
           "node_ids": {s: g.to_numpy(np.int64) for s, g in sel.groupby("stem", sort=False)["node_id"]},
           "video_fold": video_fold, "routing": routing, "code_sha": code_shas(),
           "discover_s": round(time.time() - t0, 1)}
    modes = list(dict.fromkeys(cfg["modes"]))
    book = score_modes(cfg, modes, ctx)
    state = {"summaries": {}, "work_dir": str(work)}
    for mode in modes:
        state["summaries"][mode] = finish_mode(cfg, mode, ctx, book[mode])
    state["total_s"] = round(time.time() - t0, 1)
    log(f"[kds] done in {state['total_s']} s")
    return state


if __name__ == "__main__":
    if len(sys.argv) > 1 and sys.argv[1] == "worker":
        sys.exit(worker_main(sys.argv[2:]))
    run(config_from_env())
```
